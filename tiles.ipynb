```python
def latlon_to_tile(lat, lon, zoom):
    lat_rad = math.radians(lat)
    n = 2.0 ** zoom
    x_tile = int((lon + 180.0) / 360.0 * n)
    y_tile = int((1.0 - math.log(math.tan(lat_rad) + (1 / math.cos(lat_rad))) / math.pi) / 2.0 * n)
    return x_tile, y_tile
```

```python
maps = Maps(access_token=MAPBOX_ACCESS_TOKEN)

# Starting Points
startLat = 40.018
startLon = -75.185
zoom = 16

# Ending Points
endLat = 40.0734
endLon = -75.2284

# Convert to x and y from mercator
xStart, yStart = latlon_to_tile(startLat, startLon, zoom)
xEnd, yEnd = latlon_to_tile(endLat, endLon, zoom)

xMin, xMax = min(xStart, xEnd), max(xStart, xEnd)
yMin, yMax = min(yStart, yEnd), max(yStart, yEnd)

# DON'T NEED FOR TILES (increments 1 per tile)
# # Take total size of Earth width in Meters, divide by 2^zoom, which allows iterating tile over tile
# tileSize = 40075016.68557849 / 2**zoom
# moveX = x * tileSize
# moveY = y * tileSize

for x in range(xMin, xMax + 1):
    for y in range(yMin, yMax + 1):
        response = maps.tile("mapbox.satellite", x, y, zoom, ".jpg")
        if response.status_code == 200:
            filename = f"./Tiles/Satellite/{zoom}_{x}_{y}.jpg"
            with open(filename, "wb") as output:
                output.write(response.content)
        else:
            print(f"Failed to get tile {x}, {y}")

for x in range(xMin, xMax + 1):
    for y in range(yMin, yMax + 1):
        response = maps.tile("mapbox.terrain-rgb", x, y, zoom, ".jpg")
        if response.status_code == 200:
            filename = f"./Tiles/Terrain/{zoom}_{x}_{y}.jpg"
            with open(filename, "wb") as output:
                output.write(response.content)
        else:
            print(f"Failed to get tile {x}, {y}")
```
