# Scenic Drive Rating Model with Terrain Data MVP

This notebook implements a multi-modal machine learning model that combines satellite imagery with terrain data to rate scenic potential.

**Features:**
- Downloads and processes Mapbox terrain-rgb tiles for elevation data
- Extracts terrain features: elevation, slope, roughness, aspect
- Multi-modal CNN architecture combining visual and terrain features
- Enhanced scenic scoring incorporating topographic diversity

**Data Sources:**
1. Satellite imagery (mapbox.satellite)
2. Terrain elevation data (mapbox.terrain-rgb)

````python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms, models
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, mean_absolute_error
from sklearn.preprocessing import StandardScaler
import requests
import time
import mercantile
from scipy import ndimage
from scipy.ndimage import gaussian_filter
import json
from typing import List, Tuple, Dict
import warnings
warnings.filterwarnings('ignore')

# Set device
device = torch.device('cuda' if torch.cuda.is_available() else 'mps' if torch.backends.mps.is_available() else 'cpu')
print(f"Using device: {device}")
````

## Terrain Data Download

Download terrain-rgb tiles from Mapbox. These tiles encode elevation data in RGB format:
- **Red**: High byte of elevation (256m increments)
- **Green**: Middle byte of elevation (1m increments) 
- **Blue**: Low byte of elevation (1/256m increments)
- **Formula**: `elevation = -10000 + ((R * 256 * 256 + G * 256 + B) * 0.1)`

````python
# Configuration - matching your existing tile setup
ACCESS_TOKEN = "pk.eyJ1IjoiaS1hLW4iLCJhIjoiY205M2VpNzI3MG04aTJqbjc5cWxpN2piaiJ9.Y0qHcSsfH-J8pSj6lkIdaA"
SATELLITE_TILESET = "mapbox.satellite"
TERRAIN_TILESET = "mapbox.terrain-rgb"
ZOOM = 12
TILES_DIR = "./tiles"
TERRAIN_DIR = "./terrain_tiles"
FORMAT = "pngraw"
DPR = "@2x"

# Same coordinate range as existing tiles
x_min, x_max = 1200, 1220
y_min, y_max = 1540, 1560

def download_terrain_tiles():
    """Download terrain-rgb tiles for elevation data"""
    os.makedirs(TERRAIN_DIR, exist_ok=True)
    
    session = requests.Session()
    session.params.update({"access_token": ACCESS_TOKEN})
    
    downloaded_count = 0
    
    for x in range(x_min, x_max + 1):
        for y in range(y_min, y_max + 1):
            terrain_filename = f"{ZOOM}_{x}_{y}{DPR}_terrain.png"
            terrain_path = os.path.join(TERRAIN_DIR, terrain_filename)
            
            # Skip if already downloaded
            if os.path.exists(terrain_path):
                continue
            
            url = (
                f"https://api.mapbox.com/v4/"
                f"{TERRAIN_TILESET}/{ZOOM}/{x}/{y}{DPR}.{FORMAT}"
            )
            
            try:
                resp = session.get(url, timeout=10)
                if resp.status_code == 200:
                    with open(terrain_path, "wb") as f:
                        f.write(resp.content)
                    print(f"Downloaded terrain tile {ZOOM}/{x}/{y}{DPR} → {terrain_path}")
                    downloaded_count += 1
                else:
                    print(f"Failed terrain {ZOOM}/{x}/{y}{DPR} (HTTP {resp.status_code})")
            except Exception as e:
                print(f"Error downloading {ZOOM}/{x}/{y}{DPR}: {e}")
            
            time.sleep(0.05)  # Rate limiting
    
    print(f"Downloaded {downloaded_count} new terrain tiles")
    return downloaded_count

# Download terrain tiles
download_terrain_tiles()
````

````python
def decode_terrain_rgb(terrain_img: np.ndarray) -> np.ndarray:
    """Decode terrain-rgb image to elevation in meters"""
    r, g, b = terrain_img[:, :, 0], terrain_img[:, :, 1], terrain_img[:, :, 2]
    elevation = -10000 + ((r.astype(np.float32) * 256 * 256 + 
                          g.astype(np.float32) * 256 + 
                          b.astype(np.float32)) * 0.1)
    return elevation

def calculate_terrain_features(elevation: np.ndarray) -> Dict[str, float]:
    """Calculate terrain features from elevation data"""
    # Basic elevation stats
    elevation_mean = np.mean(elevation)
    elevation_std = np.std(elevation)
    elevation_range = np.max(elevation) - np.min(elevation)
    
    # Slope calculation (gradient magnitude)
    dy, dx = np.gradient(elevation)
    slope = np.sqrt(dx**2 + dy**2)
    slope_mean = np.mean(slope)
    slope_std = np.std(slope)
    
    # Aspect (direction of slope)
    aspect = np.arctan2(dy, dx)
    aspect_variation = np.std(aspect)
    
    # Terrain roughness (local variation)
    laplacian = ndimage.laplace(elevation)
    roughness = np.std(laplacian)
    
    # Curvature features
    dxx = ndimage.sobel(dx, axis=1)
    dyy = ndimage.sobel(dy, axis=0)
    dxy = ndimage.sobel(dx, axis=0)
    
    # Avoid division by zero
    slope_safe = np.where(slope > 1e-6, slope, 1e-6)
    profile_curvature = (dxx * dx**2 + 2 * dxy * dx * dy + dyy * dy**2) / (slope_safe**3)
    profile_curvature_mean = np.mean(np.abs(profile_curvature))
    
    # Terrain Position Index (TPI)
    kernel = np.ones((5, 5)) / 25
    smoothed_elevation = ndimage.convolve(elevation, kernel, mode='constant')
    tpi = elevation - smoothed_elevation
    tpi_std = np.std(tpi)
    
    return {
        'elevation_mean': elevation_mean,
        'elevation_std': elevation_std,
        'elevation_range': elevation_range,
        'slope_mean': slope_mean,
        'slope_std': slope_std,
        'aspect_variation': aspect_variation,
        'roughness': roughness,
        'profile_curvature_mean': profile_curvature_mean,
        'tpi_std': tpi_std
    }

# Test terrain processing
test_terrain_files = [f for f in os.listdir(TERRAIN_DIR) if f.endswith('_terrain.png')]
if test_terrain_files:
    test_file = test_terrain_files[99]
    test_path = os.path.join(TERRAIN_DIR, test_file)
    
    terrain_img = np.array(Image.open(test_path))
    elevation = decode_terrain_rgb(terrain_img)
    features = calculate_terrain_features(elevation)
    
    print(f"Testing terrain processing on {test_file}:")
    print(f"Elevation range: {elevation.min():.1f}m to {elevation.max():.1f}m")
    print(f"Terrain features: {features}")
    
    # Visualize terrain data
    fig, axes = plt.subplots(1, 3, figsize=(15, 5))
    
    axes[0].imshow(terrain_img)
    axes[0].set_title('Terrain RGB')
    axes[0].axis('off')
    
    im1 = axes[1].imshow(elevation, cmap='terrain')
    axes[1].set_title('Elevation (meters)')
    axes[1].axis('off')
    plt.colorbar(im1, ax=axes[1])
    
    dy, dx = np.gradient(elevation)
    slope = np.sqrt(dx**2 + dy**2)
    im2 = axes[2].imshow(slope, cmap='hot')
    axes[2].set_title('Slope Magnitude')
    axes[2].axis('off')
    plt.colorbar(im2, ax=axes[2])
    
    plt.tight_layout()
    plt.show()
else:
    print("No terrain tiles found. Run the download cell first.")
````

## Multi-Modal Dataset

Create a dataset that combines satellite imagery with terrain features for training.

````python
class MultiModalScenicDataset(Dataset):
    def __init__(self, tiles_dir: str, terrain_dir: str, transform=None, create_labels=True):
        self.tiles_dir = tiles_dir
        self.terrain_dir = terrain_dir
        self.transform = transform
        
        # Find matching satellite and terrain tiles
        self.tile_pairs = self._find_matching_tiles()
        print(f"Found {len(self.tile_pairs)} matching satellite-terrain tile pairs")
        
        if create_labels:
            self.labels, self.terrain_features = self._generate_enhanced_scenic_scores()
            self.scaler = StandardScaler()
            self.terrain_features_normalized = self.scaler.fit_transform(self.terrain_features)
        else:
            self.labels = [0] * len(self.tile_pairs)
            self.terrain_features = np.zeros((len(self.tile_pairs), 9))
            self.terrain_features_normalized = self.terrain_features
    
    def _find_matching_tiles(self) -> List[Tuple[str, str]]:
        """Find satellite tiles that have corresponding terrain tiles"""
        satellite_files = [f for f in os.listdir(self.tiles_dir) if f.endswith('.png')]
        terrain_files = [f for f in os.listdir(self.terrain_dir) if f.endswith('_terrain.png')]
        
        pairs = []
        for sat_file in satellite_files:
            # Extract coordinates from satellite filename
            base_name = sat_file.replace('.png', '')
            terrain_file = f"{base_name}_terrain.png"
            
            if terrain_file in terrain_files:
                pairs.append((sat_file, terrain_file))
        
        return pairs
    
    def _generate_enhanced_scenic_scores(self) -> Tuple[List[float], np.ndarray]:
        """Generate scenic scores incorporating both visual and terrain features"""
        scores = []
        terrain_features_list = []
        
        for sat_file, terrain_file in self.tile_pairs:
            # Load satellite image
            sat_path = os.path.join(self.tiles_dir, sat_file)
            sat_img = np.array(Image.open(sat_path).convert('RGB'))
            
            # Load and process terrain data
            terrain_path = os.path.join(self.terrain_dir, terrain_file)
            terrain_img = np.array(Image.open(terrain_path))
            elevation = decode_terrain_rgb(terrain_img)
            terrain_features = calculate_terrain_features(elevation)
            
            # Calculate enhanced scenic score
            visual_score = self._analyze_visual_features(sat_img)
            terrain_score = self._analyze_terrain_features(terrain_features)
            
            combined_score = visual_score + terrain_score
            scores.append(min(10.0, max(0.0, combined_score)))
            
            # Store terrain features for model input
            feature_vector = [
                terrain_features['elevation_mean'],
                terrain_features['elevation_std'],
                terrain_features['elevation_range'],
                terrain_features['slope_mean'],
                terrain_features['slope_std'],
                terrain_features['aspect_variation'],
                terrain_features['roughness'],
                terrain_features['profile_curvature_mean'],
                terrain_features['tpi_std']
            ]
            terrain_features_list.append(feature_vector)
        
        return scores, np.array(terrain_features_list)
    
    def _analyze_visual_features(self, img_array: np.ndarray) -> float:
        """Analyze visual features from satellite imagery"""
        score = 3.0  # Reduced base score since terrain will add more
        h, w, _ = img_array.shape
        
        # Water detection
        blue_mask = (
            (img_array[:,:,2] > img_array[:,:,1]) & 
            (img_array[:,:,2] > img_array[:,:,0]) &
            (img_array[:,:,2] > 100)
        )
        water_ratio = np.sum(blue_mask) / (h * w)
        score += water_ratio * 2.5
        
        # Vegetation detection
        green_mask = (
            (img_array[:,:,1] > img_array[:,:,0]) & 
            (img_array[:,:,1] > img_array[:,:,2]) &
            (img_array[:,:,1] > 80)
        )
        green_ratio = np.sum(green_mask) / (h * w)
        score += green_ratio * 1.5
        
        # Visual terrain variation
        gray = np.mean(img_array, axis=2)
        visual_variation = np.std(gray) / 255.0
        score += visual_variation * 1.0
        
        # Urban penalty
        gray_brown_mask = (
            (np.abs(img_array[:,:,0] - img_array[:,:,1]) < 30) &
            (np.abs(img_array[:,:,1] - img_array[:,:,2]) < 30) &
            (np.mean(img_array, axis=2) < 150)
        )
        urban_ratio = np.sum(gray_brown_mask) / (h * w)
        score -= urban_ratio * 0.5
        
        return score
    
    def _analyze_terrain_features(self, terrain_features: Dict[str, float]) -> float:
        """Analyze terrain features for scenic scoring"""
        score = 0.0
        
        # Elevation variation adds scenic value
        elevation_range = terrain_features['elevation_range']
        elevation_std = terrain_features['elevation_std']
        
        # More elevation variation = more scenic (up to a point)
        score += min(2.0, elevation_range / 100.0)  # Max 2 points for 200m+ range
        score += min(1.5, elevation_std / 50.0)     # Max 1.5 points for 50m+ std
        
        # Moderate slopes are most scenic
        slope_mean = terrain_features['slope_mean']
        if slope_mean < 1.0:  # Very flat
            score += 0.0
        elif slope_mean < 5.0:  # Gentle hills
            score += 1.5
        elif slope_mean < 15.0:  # Moderate slopes
            score += 2.0
        else:  # Very steep
            score += 1.0
        
        # Terrain roughness adds character
        roughness = terrain_features['roughness']
        score += min(1.0, roughness / 10.0)
        
        # Curvature adds scenic interest
        curvature = terrain_features['profile_curvature_mean']
        score += min(1.0, curvature * 100.0)
        
        # Terrain position variation
        tpi_std = terrain_features['tpi_std']
        score += min(0.5, tpi_std / 20.0)
        
        return score
    
    def __len__(self):
        return len(self.tile_pairs)
    
    def __getitem__(self, idx):
        sat_file, terrain_file = self.tile_pairs[idx]
        
        # Load satellite image
        sat_path = os.path.join(self.tiles_dir, sat_file)
        image = Image.open(sat_path).convert('RGB')
        
        if self.transform:
            image = self.transform(image)
        
        # Get terrain features
        terrain_features = torch.tensor(self.terrain_features_normalized[idx], dtype=torch.float32)
        
        return {
            'image': image,
            'terrain_features': terrain_features,
            'label': torch.tensor(self.labels[idx], dtype=torch.float32),
            'sat_filename': sat_file,
            'terrain_filename': terrain_file
        }

# Data transforms
transform_train = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(0.5),
    transforms.RandomRotation(10),
    transforms.ColorJitter(brightness=0.2, contrast=0.2),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])

transform_val = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])

# Create multi-modal dataset
dataset = MultiModalScenicDataset(TILES_DIR, TERRAIN_DIR, transform=transform_val)

if len(dataset) > 0:
    print(f"Loaded {len(dataset)} tile pairs")
    print(f"Scenic score range: {min(dataset.labels):.2f} - {max(dataset.labels):.2f}")
    print(f"Mean scenic score: {np.mean(dataset.labels):.2f}")
    print(f"Terrain features shape: {dataset.terrain_features.shape}")
else:
    print("No matching tile pairs found. Make sure both satellite and terrain tiles are downloaded.")
````

## Multi-Modal CNN Architecture

Create a model that processes both satellite imagery and terrain features.

````python
class MultiModalScenicCNN(nn.Module):
    def __init__(self, terrain_features_dim=9, dropout=0.5):
        super(MultiModalScenicCNN, self).__init__()
        
        # Visual feature extractor (ResNet18 backbone)
        self.visual_backbone = models.resnet18(pretrained=True)
        
        # Remove the final classification layer
        visual_features_dim = self.visual_backbone.fc.in_features
        self.visual_backbone.fc = nn.Identity()
        
        # Terrain feature processor
        self.terrain_processor = nn.Sequential(
            nn.Linear(terrain_features_dim, 64),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(64, 32),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(32, 16),
            nn.ReLU()
        )
        
        # Fusion layer
        combined_features_dim = visual_features_dim + 16
        self.fusion_layer = nn.Sequential(
            nn.Linear(combined_features_dim, 512),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(512, 256),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(256, 128),
            nn.ReLU(),
            nn.Linear(128, 1),
            nn.Sigmoid()  # Output 0-1, will scale to 0-10
        )
        
        # Attention mechanism for feature importance
        self.visual_attention = nn.Linear(visual_features_dim, 1)
        self.terrain_attention = nn.Linear(16, 1)
        
    def forward(self, image, terrain_features):
        # Extract visual features
        visual_features = self.visual_backbone(image)
        
        # Process terrain features
        terrain_processed = self.terrain_processor(terrain_features)
        
        # Apply attention
        visual_attention = torch.sigmoid(self.visual_attention(visual_features))
        terrain_attention = torch.sigmoid(self.terrain_attention(terrain_processed))
        
        visual_weighted = visual_features * visual_attention
        terrain_weighted = terrain_processed * terrain_attention
        
        # Combine features
        combined_features = torch.cat([visual_weighted, terrain_weighted], dim=1)
        
        # Final prediction
        output = self.fusion_layer(combined_features) * 10  # Scale to 0-10
        
        return output

# Initialize model
model = MultiModalScenicCNN(terrain_features_dim=9).to(device)
print(f"Multi-modal model initialized")
print(f"Model parameters: {sum(p.numel() for p in model.parameters()):,}")

# Test forward pass
if len(dataset) > 0:
    sample = dataset[0]
    test_image = sample['image'].unsqueeze(0).to(device)
    test_terrain = sample['terrain_features'].unsqueeze(0).to(device)
    
    with torch.no_grad():
        test_output = model(test_image, test_terrain)
        print(f"Test forward pass successful. Output: {test_output.item():.2f}")
````

## Model Evaluation and Inference

Evaluate the trained model and create an inference pipeline.

````python
# Model evaluation
if len(dataset) > 0:
    model.eval()
    val_predictions = []
    val_targets = []
    
    with torch.no_grad():
        for batch in val_loader:
            images = batch['image'].to(device)
            terrain_features = batch['terrain_features'].to(device)
            labels = batch['label']
            
            outputs = model(images, terrain_features).squeeze().cpu()
            
            val_predictions.extend(outputs.numpy() if outputs.dim() > 0 else [outputs.item()])
            val_targets.extend(labels.numpy())
    
    val_predictions = np.array(val_predictions)
    val_targets = np.array(val_targets)
    
    # Calculate metrics
    mse = mean_squared_error(val_targets, val_predictions)
    mae = mean_absolute_error(val_targets, val_predictions)
    rmse = np.sqrt(mse)
    
    print(f"Multi-Modal CNN Performance:")
    print(f"MSE: {mse:.4f}")
    print(f"MAE: {mae:.4f}")
    print(f"RMSE: {rmse:.4f}")
    
    # Plot results
    plt.figure(figsize=(15, 5))
    
    plt.subplot(1, 3, 1)
    plt.scatter(val_targets, val_predictions, alpha=0.6)
    plt.plot([0, 10], [0, 10], 'r--', label='Perfect Prediction')
    plt.xlabel('True Scenic Score')
    plt.ylabel('Predicted Scenic Score')
    plt.title('Multi-Modal Predictions vs True Values')
    plt.legend()
    plt.grid(True)
    
    plt.subplot(1, 3, 2)
    residuals = val_predictions - val_targets
    plt.hist(residuals, bins=15, edgecolor='black', alpha=0.7)
    plt.xlabel('Residuals (Predicted - True)')
    plt.ylabel('Frequency')
    plt.title('Residuals Distribution')
    plt.grid(True)
    
    plt.subplot(1, 3, 3)
    errors = np.abs(residuals)
    plt.hist(errors, bins=15, edgecolor='black', alpha=0.7, color='orange')
    plt.xlabel('Absolute Error')
    plt.ylabel('Frequency')
    plt.title('Absolute Error Distribution')
    plt.grid(True)
    
    plt.tight_layout()
    plt.show()
    
    # Save the model
    model_save_path = 'multimodal_scenic_model.pth'
    torch.save({
        'model_state_dict': model.state_dict(),
        'model_architecture': 'MultiModalScenicCNN',
        'scaler_mean': dataset.scaler.mean_,
        'scaler_scale': dataset.scaler.scale_,
        'terrain_features_dim': 9,
        'metrics': {
            'mse': mse,
            'mae': mae,
            'rmse': rmse
        },
        'score_range': [0, 10]
    }, model_save_path)
    
    print(f"Multi-modal model saved to {model_save_path}")
    
else:
    print("Cannot evaluate model - no dataset available")
````

## Inference Pipeline

Create a complete inference pipeline for scoring new satellite-terrain tile pairs.

````python
class MultiModalScenicScorer:
    def __init__(self, model, transform, scaler, device):
        self.model = model
        self.transform = transform
        self.scaler = scaler
        self.device = device
        
    def score_tile_pair(self, satellite_path: str, terrain_path: str) -> float:
        """Score a satellite-terrain tile pair"""
        # Load satellite image
        sat_image = Image.open(satellite_path).convert('RGB')
        
        # Load and process terrain
        terrain_img = np.array(Image.open(terrain_path))
        elevation = decode_terrain_rgb(terrain_img)
        terrain_features = calculate_terrain_features(elevation)
        
        return self.score_inputs(sat_image, terrain_features)
    
    def score_inputs(self, sat_image: Image.Image, terrain_features: Dict[str, float]) -> float:
        """Score satellite image and terrain features"""
        self.model.eval()
        
        with torch.no_grad():
            # Transform satellite image
            img_tensor = self.transform(sat_image).unsqueeze(0).to(self.device)
            
            # Prepare terrain features
            feature_vector = np.array([
                terrain_features['elevation_mean'],
                terrain_features['elevation_std'],
                terrain_features['elevation_range'],
                terrain_features['slope_mean'],
                terrain_features['slope_std'],
                terrain_features['aspect_variation'],
                terrain_features['roughness'],
                terrain_features['profile_curvature_mean'],
                terrain_features['tpi_std']
            ]).reshape(1, -1)
            
            # Normalize terrain features
            feature_vector_normalized = self.scaler.transform(feature_vector)
            terrain_tensor = torch.tensor(feature_vector_normalized, dtype=torch.float32).to(self.device)
            
            # Get prediction
            output = self.model(img_tensor, terrain_tensor)
            score = output.squeeze().cpu().item()
            
            return max(0, min(10, score))
    
    def analyze_tile_pair(self, satellite_path: str, terrain_path: str) -> Dict:
        """Provide detailed analysis of a tile pair"""
        # Load images
        sat_image = Image.open(satellite_path).convert('RGB')
        terrain_img = np.array(Image.open(terrain_path))
        elevation = decode_terrain_rgb(terrain_img)
        
        # Calculate features
        terrain_features = calculate_terrain_features(elevation)
        
        # Get score
        score = self.score_inputs(sat_image, terrain_features)
        
        # Terrain analysis
        elev_range = terrain_features['elevation_range']
        slope_mean = terrain_features['slope_mean']
        
        if elev_range < 50:
            terrain_type = 'Flat/Plains'
        elif elev_range < 200:
            terrain_type = 'Rolling Hills'
        elif elev_range < 500:
            terrain_type = 'Hilly'
        else:
            terrain_type = 'Mountainous'
        
        if slope_mean < 2:
            slope_category = 'Gentle'
        elif slope_mean < 8:
            slope_category = 'Moderate'
        elif slope_mean < 20:
            slope_category = 'Steep'
        else:
            slope_category = 'Very Steep'
        
        return {
            'scenic_score': score,
            'elevation_stats': {
                'min': float(elevation.min()),
                'max': float(elevation.max()),
                'mean': float(elevation.mean()),
                'std': float(elevation.std())
            },
            'terrain_features': terrain_features,
            'terrain_type': terrain_type,
            'slope_category': slope_category
        }

# Create scorer and test
if len(dataset) > 0:
    scorer = MultiModalScenicScorer(model, transform_val, dataset.scaler, device)
    
    print("Testing Multi-Modal Scenic Scorer:")
    test_pairs = dataset.tile_pairs[:3]
    
    for sat_file, terrain_file in test_pairs:
        sat_path = os.path.join(TILES_DIR, sat_file)
        terrain_path = os.path.join(TERRAIN_DIR, terrain_file)
        
        analysis = scorer.analyze_tile_pair(sat_path, terrain_path)
        
        print(f"\n{sat_file}:")
        print(f"  Scenic Score: {analysis['scenic_score']:.2f}/10")
        print(f"  Terrain Type: {analysis['terrain_type']}")
        print(f"  Slope Category: {analysis['slope_category']}")
        print(f"  Elevation Range: {analysis['elevation_stats']['min']:.0f}m - {analysis['elevation_stats']['max']:.0f}m")
else:
    print("Cannot create scorer - no dataset available")
````
