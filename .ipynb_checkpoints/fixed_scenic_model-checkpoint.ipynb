```python
# Configuration - Denver to Rockies region
ACCESS_TOKEN = "pk.eyJ1IjoiaS1hLW4iLCJhIjoiY205M2VpNzI3MG04aTJqbjc5cWxpN2piaiJ9.Y0qHcSsfH-J8pSj6lkIdaA"
SATELLITE_TILESET = "mapbox.satellite"
TERRAIN_TILESET = "mapbox.terrain-rgb"
ZOOM = 14
TILES_DIR = "./improved_tiles"
TERRAIN_DIR = "./improved_terrain_tiles"
FORMAT = "pngraw"
DPR = "@2x"

def deg2num(lat_deg, lon_deg, zoom):
    import math
    lat_rad = math.radians(lat_deg)
    n = 2.0 ** zoom
    x = int((lon_deg + 180.0) / 360.0 * n)
    y = int((1.0 - math.asinh(math.tan(lat_rad)) / math.pi) / 2.0 * n)
    return (x, y)

# Denver metro to foothills region
sw_lat, sw_lon = 39.6, -105.2
ne_lat, ne_lon = 40.1, -104.8

sw_x, sw_y = deg2num(sw_lat, sw_lon, ZOOM)
ne_x, ne_y = deg2num(ne_lat, ne_lon, ZOOM)

x_min, x_max = min(sw_x, ne_x), max(sw_x, ne_x)
y_min, y_max = min(sw_y, ne_y), max(sw_y, ne_y)

print(f"Tile range: x={x_min}-{x_max}, y={y_min}-{y_max}")
print(f"Total tiles: {(x_max-x_min+1) * (y_max-y_min+1)}")
```

```python
def download_improved_tiles():
    os.makedirs(TILES_DIR, exist_ok=True)
    os.makedirs(TERRAIN_DIR, exist_ok=True)
    
    session = requests.Session()
    session.params.update({"access_token": ACCESS_TOKEN})
    
    sat_downloaded = 0
    terrain_downloaded = 0
    
    for x in range(x_min, x_max + 1):
        for y in range(y_min, y_max + 1):
            # Download satellite tile
            sat_filename = f"{ZOOM}_{x}_{y}{DPR}.png"
            sat_path = os.path.join(TILES_DIR, sat_filename)
            
            if not os.path.exists(sat_path):
                sat_url = f"https://api.mapbox.com/v4/{SATELLITE_TILESET}/{ZOOM}/{x}/{y}{DPR}.{FORMAT}"
                try:
                    resp = session.get(sat_url, timeout=10)
                    if resp.status_code == 200:
                        with open(sat_path, "wb") as f:
                            f.write(resp.content)
                        sat_downloaded += 1
                        if sat_downloaded % 10 == 0:
                            print(f"Downloaded {sat_downloaded} satellite tiles...")
                except Exception as e:
                    print(f"Error downloading satellite {ZOOM}/{x}/{y}: {e}")
            
            # Download terrain tile
            terrain_filename = f"{ZOOM}_{x}_{y}{DPR}_terrain.png"
            terrain_path = os.path.join(TERRAIN_DIR, terrain_filename)
            
            if not os.path.exists(terrain_path):
                terrain_url = f"https://api.mapbox.com/v4/{TERRAIN_TILESET}/{ZOOM}/{x}/{y}{DPR}.{FORMAT}"
                try:
                    resp = session.get(terrain_url, timeout=10)
                    if resp.status_code == 200:
                        with open(terrain_path, "wb") as f:
                            f.write(resp.content)
                        terrain_downloaded += 1
                        if terrain_downloaded % 10 == 0:
                            print(f"Downloaded {terrain_downloaded} terrain tiles...")
                except Exception as e:
                    print(f"Error downloading terrain {ZOOM}/{x}/{y}: {e}")
            
            time.sleep(0.05)
    
    print(f"Downloaded {sat_downloaded} satellite tiles, {terrain_downloaded} terrain tiles")
    return sat_downloaded, terrain_downloaded

download_improved_tiles()
```

```python
def enhanced_visual_analysis(img_array: np.ndarray) -> Dict[str, float]:
    # Ensure RGB format
    if len(img_array.shape) == 2:
        img_array = np.stack([img_array] * 3, axis=-1)
    elif len(img_array.shape) == 3 and img_array.shape[2] == 4:
        img_array = img_array[:, :, :3]
    elif len(img_array.shape) != 3 or img_array.shape[2] != 3:
        raise ValueError(f"Unsupported image format: {img_array.shape}")
    
    h, w, _ = img_array.shape
    hsv = cv2.cvtColor(img_array, cv2.COLOR_RGB2HSV)
    features = {}
    
    # Water detection - multiple methods
    blue_dominant = (
        (img_array[:,:,2] > img_array[:,:,1] + 10) & 
        (img_array[:,:,2] > img_array[:,:,0] + 10) &
        (img_array[:,:,2] > 80)
    )
    
    water_hsv = (
        (hsv[:,:,1] < 100) &
        (hsv[:,:,2] > 50) &
        (hsv[:,:,2] < 200)
    )
    
    dark_blue = (
        ((hsv[:,:,0] > 90) & (hsv[:,:,0] < 130)) &
        (hsv[:,:,1] > 50) &
        (hsv[:,:,2] > 30)
    )
    
    water_combined = blue_dominant | water_hsv | dark_blue
    features['water_ratio'] = np.sum(water_combined) / (h * w)
    
    # Vegetation detection
    green_dominant = (
        (img_array[:,:,1] > img_array[:,:,0] + 15) & 
        (img_array[:,:,1] > img_array[:,:,2] + 15) &
        (img_array[:,:,1] > 60)
    )
    
    green = img_array[:,:,1].astype(np.float32)
    red = img_array[:,:,0].astype(np.float32)
    ndvi_proxy = (green - red) / (green + red + 1e-8)
    vegetation_ndvi = ndvi_proxy > 0.1
    
    vegetation_combined = green_dominant | vegetation_ndvi
    features['vegetation_ratio'] = np.sum(vegetation_combined) / (h * w)
    
    # Snow/ice detection
    snow_mask = (
        (img_array[:,:,0] > 200) & 
        (img_array[:,:,1] > 200) & 
        (img_array[:,:,2] > 200) &
        (np.abs(img_array[:,:,0] - img_array[:,:,1]) < 20) &
        (np.abs(img_array[:,:,1] - img_array[:,:,2]) < 20)
    )
    features['snow_ratio'] = np.sum(snow_mask) / (h * w)
    
    # Rock/bare earth
    rock_mask = (
        (hsv[:,:,0] > 10) & (hsv[:,:,0] < 30) &
        (hsv[:,:,1] > 30) & (hsv[:,:,1] < 150) &
        (hsv[:,:,2] > 50) & (hsv[:,:,2] < 180)
    )
    features['rock_ratio'] = np.sum(rock_mask) / (h * w)
    
    # Urban areas
    urban_mask = (
        (np.abs(img_array[:,:,0] - img_array[:,:,1]) < 25) &
        (np.abs(img_array[:,:,1] - img_array[:,:,2]) < 25) &
        (np.mean(img_array, axis=2) > 60) &
        (np.mean(img_array, axis=2) < 180)
    )
    features['urban_ratio'] = np.sum(urban_mask) / (h * w)
    
    # Texture analysis
    gray = cv2.cvtColor(img_array, cv2.COLOR_RGB2GRAY)
    edges = cv2.Canny(gray, 50, 150)
    features['edge_density'] = np.sum(edges > 0) / (h * w)
    
    kernel = cv2.getGaussianKernel(9, 2)
    kernel = kernel @ kernel.T
    mean_filtered = cv2.filter2D(gray.astype(np.float32), -1, kernel)
    sq_filtered = cv2.filter2D((gray.astype(np.float32))**2, -1, kernel)
    local_variance = sq_filtered - mean_filtered**2
    features['texture_variation'] = np.mean(np.sqrt(np.maximum(local_variance, 0)))
    
    # Color diversity
    try:
        if h * w > 50000:
            step = max(1, int(np.sqrt(h * w / 50000)))
            img_sample = img_array[::step, ::step, :]
        else:
            img_sample = img_array
        
        img_quantized = (img_sample // 16) * 16
        img_flat = img_quantized.reshape(-1, 3)
        unique_colors = len(np.unique(img_flat.view(np.dtype((np.void, img_flat.dtype.itemsize * 3)))))
        features['color_diversity'] = min(1.0, unique_colors / (img_sample.shape[0] * img_sample.shape[1]))
    except:
        features['color_diversity'] = min(1.0, np.std(img_array) / 255.0)
    
    return features

# Test the function
test_files = [f for f in os.listdir(TILES_DIR) if f.endswith('.png')]
if test_files:
    test_file = test_files[0]
    test_img = np.array(Image.open(os.path.join(TILES_DIR, test_file)).convert('RGB'))
    
    print(f"Testing visual analysis on {test_file}")
    print(f"Image shape: {test_img.shape}")
    
    features = enhanced_visual_analysis(test_img)
    
    print("\nVisual features:")
    for feature, value in features.items():
        print(f"  {feature}: {value:.4f}")
        
    # Quick visualization
    plt.figure(figsize=(12, 4))
    
    plt.subplot(1, 3, 1)
    plt.imshow(test_img)
    plt.title('Original Image')
    plt.axis('off')
    
    plt.subplot(1, 3, 2)
    hsv = cv2.cvtColor(test_img, cv2.COLOR_RGB2HSV)
    water_mask = (
        (test_img[:,:,2] > test_img[:,:,1] + 10) & 
        (test_img[:,:,2] > test_img[:,:,0] + 10) &
        (test_img[:,:,2] > 80)
    )
    plt.imshow(water_mask, cmap='Blues')
    plt.title(f'Water Detection ({features["water_ratio"]:.3f})')
    plt.axis('off')
    
    plt.subplot(1, 3, 3)
    veg_mask = (
        (test_img[:,:,1] > test_img[:,:,0] + 15) & 
        (test_img[:,:,1] > test_img[:,:,2] + 15) &
        (test_img[:,:,1] > 60)
    )
    plt.imshow(veg_mask, cmap='Greens')
    plt.title(f'Vegetation Detection ({features["vegetation_ratio"]:.3f})')
    plt.axis('off')
    
    plt.tight_layout()
    plt.show()
else:
    print("No tiles found. Run download first.")
```

```python
def decode_terrain_rgb(terrain_img: np.ndarray) -> np.ndarray:
    r, g, b = terrain_img[:, :, 0], terrain_img[:, :, 1], terrain_img[:, :, 2]
    elevation = -10000 + ((r.astype(np.float32) * 256 * 256 + 
                          g.astype(np.float32) * 256 + 
                          b.astype(np.float32)) * 0.1)
    return elevation

def calculate_terrain_features(elevation: np.ndarray) -> Dict[str, float]:
    elevation_mean = np.mean(elevation)
    elevation_std = np.std(elevation)
    elevation_range = np.max(elevation) - np.min(elevation)
    
    dy, dx = np.gradient(elevation)
    slope = np.sqrt(dx**2 + dy**2)
    slope_mean = np.mean(slope)
    slope_std = np.std(slope)
    
    aspect = np.arctan2(dy, dx)
    aspect_variation = np.std(aspect)
    
    laplacian = ndimage.laplace(elevation)
    roughness = np.std(laplacian)
    
    kernel = np.ones((5, 5)) / 25
    smoothed_elevation = ndimage.convolve(elevation, kernel, mode='constant')
    tpi = elevation - smoothed_elevation
    tpi_std = np.std(tpi)
    
    return {
        'elevation_mean': elevation_mean,
        'elevation_std': elevation_std,
        'elevation_range': elevation_range,
        'slope_mean': slope_mean,
        'slope_std': slope_std,
        'aspect_variation': aspect_variation,
        'roughness': roughness,
        'tpi_std': tpi_std
    }

# Test terrain processing
terrain_files = [f for f in os.listdir(TERRAIN_DIR) if f.endswith('_terrain.png')]
if terrain_files:
    test_terrain = terrain_files[0]
    terrain_img = np.array(Image.open(os.path.join(TERRAIN_DIR, test_terrain)))
    elevation = decode_terrain_rgb(terrain_img)
    terrain_features = calculate_terrain_features(elevation)
    
    print(f"Testing terrain analysis on {test_terrain}")
    print(f"Elevation range: {elevation.min():.1f}m to {elevation.max():.1f}m")
    
    print("\nTerrain features:")
    for feature, value in terrain_features.items():
        print(f"  {feature}: {value:.4f}")
        
    # Quick visualization
    plt.figure(figsize=(12, 4))
    
    plt.subplot(1, 3, 1)
    plt.imshow(terrain_img)
    plt.title('Terrain RGB')
    plt.axis('off')
    
    plt.subplot(1, 3, 2)
    plt.imshow(elevation, cmap='terrain')
    plt.title(f'Elevation ({elevation.min():.0f}-{elevation.max():.0f}m)')
    plt.colorbar()
    plt.axis('off')
    
    plt.subplot(1, 3, 3)
    dy, dx = np.gradient(elevation)
    slope = np.sqrt(dx**2 + dy**2)
    plt.imshow(slope, cmap='hot')
    plt.title(f'Slope (mean: {terrain_features["slope_mean"]:.2f})')
    plt.colorbar()
    plt.axis('off')
    
    plt.tight_layout()
    plt.show()
else:
    print("No terrain files found. Run download first.")
```

```python
class ScenicDataset(Dataset):
    def __init__(self, tiles_dir: str, terrain_dir: str, transform=None):
        self.tiles_dir = tiles_dir
        self.terrain_dir = terrain_dir
        self.transform = transform
        
        self.tile_pairs = self._find_matching_tiles()
        print(f"Found {len(self.tile_pairs)} matching tile pairs")
        
        if len(self.tile_pairs) > 0:
            self.labels, self.visual_features, self.terrain_features = self._generate_scores()
            
            self.visual_scaler = StandardScaler()
            self.terrain_scaler = StandardScaler()
            
            self.visual_features_normalized = self.visual_scaler.fit_transform(self.visual_features)
            self.terrain_features_normalized = self.terrain_scaler.fit_transform(self.terrain_features)
        else:
            self.labels = []
            self.visual_features = np.array([])
            self.terrain_features = np.array([])
    
    def _find_matching_tiles(self):
        if not os.path.exists(self.tiles_dir) or not os.path.exists(self.terrain_dir):
            return []
            
        satellite_files = [f for f in os.listdir(self.tiles_dir) if f.endswith('.png')]
        terrain_files = [f for f in os.listdir(self.terrain_dir) if f.endswith('_terrain.png')]
        
        pairs = []
        for sat_file in satellite_files:
            base_name = sat_file.replace('.png', '')
            terrain_file = f"{base_name}_terrain.png"
            
            if terrain_file in terrain_files:
                pairs.append((sat_file, terrain_file))
        
        return pairs
    
    def _generate_scores(self):
        scores = []
        visual_features_list = []
        terrain_features_list = []
        
        for i, (sat_file, terrain_file) in enumerate(self.tile_pairs):
            if i % 50 == 0:
                print(f"Processing {i}/{len(self.tile_pairs)} tiles...")
                
            try:
                # Load satellite image
                sat_path = os.path.join(self.tiles_dir, sat_file)
                sat_img = np.array(Image.open(sat_path).convert('RGB'))
                visual_features = enhanced_visual_analysis(sat_img)
                
                # Load terrain
                terrain_path = os.path.join(self.terrain_dir, terrain_file)
                terrain_img = np.array(Image.open(terrain_path))
                elevation = decode_terrain_rgb(terrain_img)
                terrain_features = calculate_terrain_features(elevation)
                
                # Calculate scenic score
                visual_score = self._calculate_visual_score(visual_features)
                terrain_score = self._calculate_terrain_score(terrain_features)
                
                # 60% visual, 40% terrain weighting
                combined_score = 0.6 * visual_score + 0.4 * terrain_score
                scores.append(min(10.0, max(0.0, combined_score)))
                
                # Store features
                visual_vector = [
                    visual_features['water_ratio'],
                    visual_features['vegetation_ratio'],
                    visual_features['snow_ratio'],
                    visual_features['rock_ratio'],
                    visual_features['urban_ratio'],
                    visual_features['edge_density'],
                    visual_features['texture_variation'],
                    visual_features['color_diversity']
                ]
                visual_features_list.append(visual_vector)
                
                terrain_vector = [
                    terrain_features['elevation_mean'],
                    terrain_features['elevation_std'],
                    terrain_features['elevation_range'],
                    terrain_features['slope_mean'],
                    terrain_features['slope_std'],
                    terrain_features['aspect_variation'],
                    terrain_features['roughness'],
                    terrain_features['tpi_std']
                ]
                terrain_features_list.append(terrain_vector)
                
            except Exception as e:
                print(f"Error processing {sat_file}: {e}")
                scores.append(5.0)
                visual_features_list.append([0.0] * 8)
                terrain_features_list.append([0.0] * 8)
        
        return scores, np.array(visual_features_list), np.array(terrain_features_list)
    
    def _calculate_visual_score(self, features):
        score = 2.0
        
        # Water is highly scenic
        score += min(3.0, features['water_ratio'] * 20)
        
        # Vegetation adds beauty
        score += min(2.0, features['vegetation_ratio'] * 8)
        
        # Snow can be scenic
        score += min(1.5, features['snow_ratio'] * 10)
        
        # Rock formations
        score += min(1.0, features['rock_ratio'] * 5)
        
        # Edge density indicates features
        score += min(1.5, features['edge_density'] * 15)
        
        # Texture variation
        score += min(1.0, features['texture_variation'] / 50)
        
        # Urban penalty
        score -= features['urban_ratio'] * 3
        
        return max(0.0, score)
    
    def _calculate_terrain_score(self, features):
        score = 1.0
        
        elev_range = features['elevation_range']
        if elev_range > 500:
            score += 3.0
        elif elev_range > 200:
            score += 2.0
        elif elev_range > 100:
            score += 1.0
        
        slope_mean = features['slope_mean']
        if 2 < slope_mean < 15:
            score += 2.0
        elif 1 < slope_mean < 25:
            score += 1.0
        
        score += min(1.5, features['roughness'] / 20)
        score += min(1.0, features['tpi_std'] / 20)
        
        return max(0.0, score)
    
    def __len__(self):
        return len(self.tile_pairs)
    
    def __getitem__(self, idx):
        sat_file, terrain_file = self.tile_pairs[idx]
        
        sat_path = os.path.join(self.tiles_dir, sat_file)
        image = Image.open(sat_path).convert('RGB')
        
        if self.transform:
            image = self.transform(image)
        
        return {
            'image': image,
            'visual_features': torch.tensor(self.visual_features_normalized[idx], dtype=torch.float32),
            'terrain_features': torch.tensor(self.terrain_features_normalized[idx], dtype=torch.float32),
            'label': torch.tensor(self.labels[idx], dtype=torch.float32),
            'filename': sat_file
        }

# Create dataset
transform_val = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])

dataset = ScenicDataset(TILES_DIR, TERRAIN_DIR, transform=transform_val)

if len(dataset) > 0:
    print(f"\nDataset created with {len(dataset)} samples")
    print(f"Scenic scores range: {min(dataset.labels):.2f} - {max(dataset.labels):.2f}")
    print(f"Mean scenic score: {np.mean(dataset.labels):.2f}")
    print(f"Visual features shape: {dataset.visual_features.shape}")
    print(f"Terrain features shape: {dataset.terrain_features.shape}")
    
    plt.figure(figsize=(10, 4))
    
    plt.subplot(1, 2, 1)
    plt.hist(dataset.labels, bins=20, edgecolor='black', alpha=0.7)
    plt.xlabel('Scenic Score')
    plt.ylabel('Frequency')
    plt.title('Scenic Score Distribution')
    plt.grid(True, alpha=0.3)
    
    plt.subplot(1, 2, 2)
    water_ratios = dataset.visual_features[:, 0]
    plt.scatter(water_ratios, dataset.labels, alpha=0.6)
    plt.xlabel('Water Ratio')
    plt.ylabel('Scenic Score')
    plt.title('Water vs Scenic Score')
    plt.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.show()
else:
    print("No dataset created. Make sure tiles are downloaded.")
```
