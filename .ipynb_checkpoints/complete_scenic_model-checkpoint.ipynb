# Complete Scenic Drive Rating Model

This notebook combines the enhanced data collection with full model training capabilities.

**Key Features:**
- **Enhanced Region**: Denver-Rockies area for diverse terrain (mountains, plains, foothills)
- **Higher Resolution**: Zoom level 14 for detailed analysis
- **Advanced Visual Features**: Multi-color space water/vegetation detection
- **HuggingFace Integration**: Pre-trained vision transformer with transfer learning
- **Balanced Scoring**: 60% visual, 40% terrain weighting
- **Complete Training**: Full model training with validation and evaluation

```python
def download_tiles():
    """Download satellite and terrain tiles"""
    os.makedirs(TILES_DIR, exist_ok=True)
    os.makedirs(TERRAIN_DIR, exist_ok=True)
    
    session = requests.Session()
    session.params.update({"access_token": ACCESS_TOKEN})
    
    sat_downloaded = 0
    terrain_downloaded = 0
    
    for x in range(x_min, x_max + 1):
        for y in range(y_min, y_max + 1):
            # Download satellite tile
            sat_filename = f"{ZOOM}_{x}_{y}{DPR}.png"
            sat_path = os.path.join(TILES_DIR, sat_filename)
            
            if not os.path.exists(sat_path):
                sat_url = f"https://api.mapbox.com/v4/{SATELLITE_TILESET}/{ZOOM}/{x}/{y}{DPR}.{FORMAT}"
                try:
                    resp = session.get(sat_url, timeout=10)
                    if resp.status_code == 200:
                        with open(sat_path, "wb") as f:
                            f.write(resp.content)
                        sat_downloaded += 1
                        if sat_downloaded % 10 == 0:
                            print(f"Downloaded {sat_downloaded} satellite tiles...")
                except Exception as e:
                    print(f"Error downloading satellite {ZOOM}/{x}/{y}: {e}")
            
            # Download terrain tile
            terrain_filename = f"{ZOOM}_{x}_{y}{DPR}_terrain.png"
            terrain_path = os.path.join(TERRAIN_DIR, terrain_filename)
            
            if not os.path.exists(terrain_path):
                terrain_url = f"https://api.mapbox.com/v4/{TERRAIN_TILESET}/{ZOOM}/{x}/{y}{DPR}.{FORMAT}"
                try:
                    resp = session.get(terrain_url, timeout=10)
                    if resp.status_code == 200:
                        with open(terrain_path, "wb") as f:
                            f.write(resp.content)
                        terrain_downloaded += 1
                        if terrain_downloaded % 10 == 0:
                            print(f"Downloaded {terrain_downloaded} terrain tiles...")
                except Exception as e:
                    print(f"Error downloading terrain {ZOOM}/{x}/{y}: {e}")
            
            time.sleep(0.05)  # Rate limiting
    
    print(f"Downloaded {sat_downloaded} satellite tiles, {terrain_downloaded} terrain tiles")
    return sat_downloaded, terrain_downloaded

# Download tiles
download_tiles()
```

```python
def enhanced_visual_analysis(img_array: np.ndarray) -> Dict[str, float]:
    """Enhanced visual feature detection with multi-color space analysis"""
    # Ensure RGB format
    if len(img_array.shape) == 2:
        img_array = np.stack([img_array] * 3, axis=-1)
    elif len(img_array.shape) == 3 and img_array.shape[2] == 4:
        img_array = img_array[:, :, :3]
    elif len(img_array.shape) != 3 or img_array.shape[2] != 3:
        raise ValueError(f"Unsupported image format: {img_array.shape}")
    
    h, w, _ = img_array.shape
    features = {}
    
    if CV2_AVAILABLE:
        hsv = cv2.cvtColor(img_array, cv2.COLOR_RGB2HSV)
        
        # Water detection - multiple methods
        blue_dominant = (
            (img_array[:,:,2] > img_array[:,:,1] + 10) & 
            (img_array[:,:,2] > img_array[:,:,0] + 10) &
            (img_array[:,:,2] > 80)
        )
        
        water_hsv = (
            (hsv[:,:,1] < 100) &
            (hsv[:,:,2] > 50) &
            (hsv[:,:,2] < 200)
        )
        
        dark_blue = (
            ((hsv[:,:,0] > 90) & (hsv[:,:,0] < 130)) &
            (hsv[:,:,1] > 50) &
            (hsv[:,:,2] > 30)
        )
        
        water_combined = blue_dominant | water_hsv | dark_blue
        features['water_ratio'] = np.sum(water_combined) / (h * w)
        
        # Vegetation detection
        green_dominant = (
            (img_array[:,:,1] > img_array[:,:,0] + 15) & 
            (img_array[:,:,1] > img_array[:,:,2] + 15) &
            (img_array[:,:,1] > 60)
        )
        
        green = img_array[:,:,1].astype(np.float32)
        red = img_array[:,:,0].astype(np.float32)
        ndvi_proxy = (green - red) / (green + red + 1e-8)
        vegetation_ndvi = ndvi_proxy > 0.1
        
        vegetation_combined = green_dominant | vegetation_ndvi
        features['vegetation_ratio'] = np.sum(vegetation_combined) / (h * w)
        
        # Texture analysis with OpenCV
        gray = cv2.cvtColor(img_array, cv2.COLOR_RGB2GRAY)
        edges = cv2.Canny(gray, 50, 150)
        features['edge_density'] = np.sum(edges > 0) / (h * w)
        
        # Local texture variation
        kernel = cv2.getGaussianKernel(9, 2)
        kernel = kernel @ kernel.T
        mean_filtered = cv2.filter2D(gray.astype(np.float32), -1, kernel)
        sq_filtered = cv2.filter2D((gray.astype(np.float32))**2, -1, kernel)
        local_variance = sq_filtered - mean_filtered**2
        features['texture_variation'] = np.mean(np.sqrt(np.maximum(local_variance, 0)))
    else:
        # Fallback without OpenCV
        blue_dominant = (
            (img_array[:,:,2] > img_array[:,:,1] + 10) & 
            (img_array[:,:,2] > img_array[:,:,0] + 10) &
            (img_array[:,:,2] > 80)
        )
        
        green_dominant = (
            (img_array[:,:,1] > img_array[:,:,0] + 15) & 
            (img_array[:,:,1] > img_array[:,:,2] + 15) &
            (img_array[:,:,1] > 60)
        )
        
        features['water_ratio'] = np.sum(blue_dominant) / (h * w)
        features['vegetation_ratio'] = np.sum(green_dominant) / (h * w)
        features['edge_density'] = 0.1  # Default value
        features['texture_variation'] = np.std(img_array)
    
    # Snow/ice detection
    snow_mask = (
        (img_array[:,:,0] > 200) & 
        (img_array[:,:,1] > 200) & 
        (img_array[:,:,2] > 200) &
        (np.abs(img_array[:,:,0] - img_array[:,:,1]) < 20) &
        (np.abs(img_array[:,:,1] - img_array[:,:,2]) < 20)
    )
    features['snow_ratio'] = np.sum(snow_mask) / (h * w)
    
    # Rock/bare earth (approximation)
    rock_mask = (
        (img_array[:,:,0] > 100) & (img_array[:,:,0] < 180) &
        (img_array[:,:,1] > 80) & (img_array[:,:,1] < 160) &
        (img_array[:,:,2] > 60) & (img_array[:,:,2] < 140)
    )
    features['rock_ratio'] = np.sum(rock_mask) / (h * w)
    
    # Urban areas
    urban_mask = (
        (np.abs(img_array[:,:,0] - img_array[:,:,1]) < 25) &
        (np.abs(img_array[:,:,1] - img_array[:,:,2]) < 25) &
        (np.mean(img_array, axis=2) > 60) &
        (np.mean(img_array, axis=2) < 180)
    )
    features['urban_ratio'] = np.sum(urban_mask) / (h * w)
    
    # Color diversity
    try:
        if h * w > 50000:
            step = max(1, int(np.sqrt(h * w / 50000)))
            img_sample = img_array[::step, ::step, :]
        else:
            img_sample = img_array
        
        img_quantized = (img_sample // 16) * 16
        img_flat = img_quantized.reshape(-1, 3)
        unique_colors = len(np.unique(img_flat.view(np.dtype((np.void, img_flat.dtype.itemsize * 3)))))
        features['color_diversity'] = min(1.0, unique_colors / (img_sample.shape[0] * img_sample.shape[1]))
    except:
        features['color_diversity'] = min(1.0, np.std(img_array) / 255.0)
    
    return features

# Test the enhanced visual analysis
test_files = [f for f in os.listdir(TILES_DIR) if f.endswith('.png')]
if test_files:
    test_file = test_files[0]
    test_img = np.array(Image.open(os.path.join(TILES_DIR, test_file)).convert('RGB'))
    
    print(f"Testing visual analysis on {test_file}")
    print(f"Image shape: {test_img.shape}")
    
    features = enhanced_visual_analysis(test_img)
    
    print("\nVisual features:")
    for feature, value in features.items():
        print(f"  {feature}: {value:.4f}")
else:
    print("No tiles found yet.")
```

```python
def decode_terrain_rgb(terrain_img: np.ndarray) -> np.ndarray:
    """Decode terrain-rgb image to elevation in meters"""
    r, g, b = terrain_img[:, :, 0], terrain_img[:, :, 1], terrain_img[:, :, 2]
    elevation = -10000 + ((r.astype(np.float32) * 256 * 256 + 
                          g.astype(np.float32) * 256 + 
                          b.astype(np.float32)) * 0.1)
    return elevation

def calculate_terrain_features(elevation: np.ndarray) -> Dict[str, float]:
    """Calculate comprehensive terrain features"""
    elevation_mean = np.mean(elevation)
    elevation_std = np.std(elevation)
    elevation_range = np.max(elevation) - np.min(elevation)
    
    dy, dx = np.gradient(elevation)
    slope = np.sqrt(dx**2 + dy**2)
    slope_mean = np.mean(slope)
    slope_std = np.std(slope)
    
    aspect = np.arctan2(dy, dx)
    aspect_variation = np.std(aspect)
    
    laplacian = ndimage.laplace(elevation)
    roughness = np.std(laplacian)
    
    kernel = np.ones((5, 5)) / 25
    smoothed_elevation = ndimage.convolve(elevation, kernel, mode='constant')
    tpi = elevation - smoothed_elevation
    tpi_std = np.std(tpi)
    
    return {
        'elevation_mean': elevation_mean,
        'elevation_std': elevation_std,
        'elevation_range': elevation_range,
        'slope_mean': slope_mean,
        'slope_std': slope_std,
        'aspect_variation': aspect_variation,
        'roughness': roughness,
        'tpi_std': tpi_std
    }

# Test terrain processing
terrain_files = [f for f in os.listdir(TERRAIN_DIR) if f.endswith('_terrain.png')]
if terrain_files:
    test_terrain = terrain_files[0]
    terrain_img = np.array(Image.open(os.path.join(TERRAIN_DIR, test_terrain)))
    elevation = decode_terrain_rgb(terrain_img)
    terrain_features = calculate_terrain_features(elevation)
    
    print(f"Testing terrain analysis on {test_terrain}")
    print(f"Elevation range: {elevation.min():.1f}m to {elevation.max():.1f}m")
    
    print("\nTerrain features:")
    for feature, value in terrain_features.items():
        print(f"  {feature}: {value:.4f}")
else:
    print("No terrain files found yet.")
```

```python
# HuggingFace Vision Model (if available)
if HUGGINGFACE_AVAILABLE:
    class HuggingFaceVisualEncoder(nn.Module):
        def __init__(self, model_name="microsoft/swin-tiny-patch4-window7-224"):
            super().__init__()
            
            try:
                self.config = AutoConfig.from_pretrained(model_name)
                self.model = AutoModel.from_pretrained(model_name)
                self.processor = AutoImageProcessor.from_pretrained(model_name)
                
                self.feature_dim = self.config.hidden_size
                
                # Freeze backbone
                for param in self.model.parameters():
                    param.requires_grad = False
                    
                # Trainable projection
                self.projection = nn.Sequential(
                    nn.Linear(self.feature_dim, 512),
                    nn.ReLU(),
                    nn.Dropout(0.3),
                    nn.Linear(512, 256)
                )
                
                self.available = True
            except Exception as e:
                print(f"HuggingFace model failed to load: {e}")
                self.available = False
                
        def forward(self, pixel_values):
            if not self.available:
                return torch.zeros(pixel_values.size(0), 256, device=pixel_values.device)
                
            outputs = self.model(pixel_values=pixel_values)
            
            if hasattr(outputs, 'pooler_output') and outputs.pooler_output is not None:
                features = outputs.pooler_output
            else:
                features = outputs.last_hidden_state.mean(dim=1)
            
            return self.projection(features)

# Advanced Multi-Modal Model
class CompleteScenicCNN(nn.Module):
    def __init__(self, visual_features_dim=8, terrain_features_dim=8, use_huggingface=HUGGINGFACE_AVAILABLE):
        super().__init__()
        
        self.use_huggingface = use_huggingface and HUGGINGFACE_AVAILABLE
        
        if self.use_huggingface:
            try:
                self.visual_encoder = HuggingFaceVisualEncoder()
                if self.visual_encoder.available:
                    visual_feature_dim = 256
                else:
                    self.use_huggingface = False
            except:
                self.use_huggingface = False
        
        if not self.use_huggingface:
            # Traditional CNN backbone
            self.visual_backbone = models.efficientnet_b0(pretrained=True)
            visual_feature_dim = self.visual_backbone.classifier[1].in_features
            self.visual_backbone.classifier = nn.Identity()
        
        # Feature processors
        self.visual_features_processor = nn.Sequential(
            nn.Linear(visual_features_dim, 32),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(32, 16)
        )
        
        self.terrain_processor = nn.Sequential(
            nn.Linear(terrain_features_dim, 32),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(32, 16)
        )
        
        # Fusion layers
        fusion_input_dim = visual_feature_dim + 16 + 16
        
        self.fusion_layers = nn.Sequential(
            nn.Linear(fusion_input_dim, 512),
            nn.ReLU(),
            nn.Dropout(0.4),
            nn.Linear(512, 256),
            nn.ReLU(),
            nn.Dropout(0.4),
            nn.Linear(256, 128),
            nn.ReLU(),
            nn.Linear(128, 64),
            nn.ReLU(),
            nn.Linear(64, 1),
            nn.Sigmoid()
        )
        
    def forward(self, images, visual_features, terrain_features):
        # Process images
        if self.use_huggingface:
            visual_encoded = self.visual_encoder(images)
        else:
            visual_encoded = self.visual_backbone(images)
        
        # Process features
        visual_features_processed = self.visual_features_processor(visual_features)
        terrain_processed = self.terrain_processor(terrain_features)
        
        # Combine features
        combined_features = torch.cat([
            visual_encoded,
            visual_features_processed,
            terrain_processed
        ], dim=1)
        
        # Final prediction (0-10 scale)
        output = self.fusion_layers(combined_features) * 10
        
        return output

# Initialize model
model = CompleteScenicCNN(visual_features_dim=8, terrain_features_dim=8)
model = model.to(device)
print(f"Model initialized with {'HuggingFace' if model.use_huggingface else 'traditional CNN'} visual encoder")
print(f"Total parameters: {sum(p.numel() for p in model.parameters()):,}")
print(f"Trainable parameters: {sum(p.numel() for p in model.parameters() if p.requires_grad):,}")
```

```python
class CompleteScenicDataset(Dataset):
    def __init__(self, tiles_dir: str, terrain_dir: str, transform=None):
        self.tiles_dir = tiles_dir
        self.terrain_dir = terrain_dir
        self.transform = transform
        
        self.tile_pairs = self._find_matching_tiles()
        print(f"Found {len(self.tile_pairs)} matching tile pairs")
        
        if len(self.tile_pairs) > 0:
            self.labels, self.visual_features, self.terrain_features = self._generate_scores()
            
            self.visual_scaler = StandardScaler()
            self.terrain_scaler = StandardScaler()
            
            self.visual_features_normalized = self.visual_scaler.fit_transform(self.visual_features)
            self.terrain_features_normalized = self.terrain_scaler.fit_transform(self.terrain_features)
        else:
            self.labels = []
            self.visual_features = np.array([])
            self.terrain_features = np.array([])
    
    def _find_matching_tiles(self):
        if not os.path.exists(self.tiles_dir) or not os.path.exists(self.terrain_dir):
            return []
            
        satellite_files = [f for f in os.listdir(self.tiles_dir) if f.endswith('.png')]
        terrain_files = [f for f in os.listdir(self.terrain_dir) if f.endswith('_terrain.png')]
        
        pairs = []
        for sat_file in satellite_files:
            base_name = sat_file.replace('.png', '')
            terrain_file = f"{base_name}_terrain.png"
            
            if terrain_file in terrain_files:
                pairs.append((sat_file, terrain_file))
        
        return pairs
    
    def _generate_scores(self):
        scores = []
        visual_features_list = []
        terrain_features_list = []
        
        for i, (sat_file, terrain_file) in enumerate(self.tile_pairs):
            if i % 50 == 0:
                print(f"Processing {i}/{len(self.tile_pairs)} tiles...")
                
            try:
                # Load satellite image
                sat_path = os.path.join(self.tiles_dir, sat_file)
                sat_img = np.array(Image.open(sat_path).convert('RGB'))
                visual_features = enhanced_visual_analysis(sat_img)
                
                # Load terrain
                terrain_path = os.path.join(self.terrain_dir, terrain_file)
                terrain_img = np.array(Image.open(terrain_path))
                elevation = decode_terrain_rgb(terrain_img)
                terrain_features = calculate_terrain_features(elevation)
                
                # Calculate scenic score (60% visual, 40% terrain)
                visual_score = self._calculate_visual_score(visual_features)
                terrain_score = self._calculate_terrain_score(terrain_features)
                
                combined_score = 0.6 * visual_score + 0.4 * terrain_score
                scores.append(min(10.0, max(0.0, combined_score)))
                
                # Store features
                visual_vector = [
                    visual_features['water_ratio'],
                    visual_features['vegetation_ratio'],
                    visual_features['snow_ratio'],
                    visual_features['rock_ratio'],
                    visual_features['urban_ratio'],
                    visual_features['edge_density'],
                    visual_features['texture_variation'],
                    visual_features['color_diversity']
                ]
                visual_features_list.append(visual_vector)
                
                terrain_vector = [
                    terrain_features['elevation_mean'],
                    terrain_features['elevation_std'],
                    terrain_features['elevation_range'],
                    terrain_features['slope_mean'],
                    terrain_features['slope_std'],
                    terrain_features['aspect_variation'],
                    terrain_features['roughness'],
                    terrain_features['tpi_std']
                ]
                terrain_features_list.append(terrain_vector)
                
            except Exception as e:
                print(f"Error processing {sat_file}: {e}")
                scores.append(5.0)
                visual_features_list.append([0.0] * 8)
                terrain_features_list.append([0.0] * 8)
        
        return scores, np.array(visual_features_list), np.array(terrain_features_list)
    
    def _calculate_visual_score(self, features):
        score = 2.0
        
        # Water is highly scenic
        score += min(3.0, features['water_ratio'] * 20)
        
        # Vegetation adds beauty
        score += min(2.0, features['vegetation_ratio'] * 8)
        
        # Snow can be scenic
        score += min(1.5, features['snow_ratio'] * 10)
        
        # Rock formations
        score += min(1.0, features['rock_ratio'] * 5)
        
        # Edge density indicates features
        score += min(1.5, features['edge_density'] * 15)
        
        # Texture variation
        score += min(1.0, features['texture_variation'] / 50)
        
        # Urban penalty
        score -= features['urban_ratio'] * 3
        
        return max(0.0, score)
    
    def _calculate_terrain_score(self, features):
        score = 1.0
        
        elev_range = features['elevation_range']
        if elev_range > 500:
            score += 3.0
        elif elev_range > 200:
            score += 2.0
        elif elev_range > 100:
            score += 1.0
        
        slope_mean = features['slope_mean']
        if 2 < slope_mean < 15:
            score += 2.0
        elif 1 < slope_mean < 25:
            score += 1.0
        
        score += min(1.5, features['roughness'] / 20)
        score += min(1.0, features['tpi_std'] / 20)
        
        return max(0.0, score)
    
    def __len__(self):
        return len(self.tile_pairs)
    
    def __getitem__(self, idx):
        sat_file, terrain_file = self.tile_pairs[idx]
        
        sat_path = os.path.join(self.tiles_dir, sat_file)
        image = Image.open(sat_path).convert('RGB')
        
        if self.transform:
            image = self.transform(image)
        
        return {
            'image': image,
            'visual_features': torch.tensor(self.visual_features_normalized[idx], dtype=torch.float32),
            'terrain_features': torch.tensor(self.terrain_features_normalized[idx], dtype=torch.float32),
            'label': torch.tensor(self.labels[idx], dtype=torch.float32),
            'filename': sat_file
        }

# Create dataset with transforms
transform_train = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(0.3),
    transforms.RandomRotation(15),
    transforms.ColorJitter(brightness=0.3, contrast=0.3, saturation=0.2, hue=0.1),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])

transform_val = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])

# Create dataset
dataset = CompleteScenicDataset(TILES_DIR, TERRAIN_DIR, transform=transform_val)

if len(dataset) > 0:
    print(f"\nDataset created with {len(dataset)} samples")
    print(f"Scenic scores range: {min(dataset.labels):.2f} - {max(dataset.labels):.2f}")
    print(f"Mean scenic score: {np.mean(dataset.labels):.2f}")
    print(f"Visual features shape: {dataset.visual_features.shape}")
    print(f"Terrain features shape: {dataset.terrain_features.shape}")
else:
    print("No dataset created. Make sure tiles are downloaded.")
```

```python
def train_model(model, train_loader, val_loader, num_epochs=20, lr=0.001):
    """Train the scenic rating model"""
    criterion = nn.MSELoss()
    optimizer = optim.Adam(model.parameters(), lr=lr, weight_decay=1e-4)
    scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', factor=0.5, patience=5)
    
    train_losses = []
    val_losses = []
    
    best_val_loss = float('inf')
    
    for epoch in range(num_epochs):
        # Training phase
        model.train()
        train_loss = 0.0
        train_samples = 0
        
        for batch in train_loader:
            images = batch['image'].to(device)
            visual_features = batch['visual_features'].to(device)
            terrain_features = batch['terrain_features'].to(device)
            labels = batch['label'].to(device)
            
            optimizer.zero_grad()
            
            outputs = model(images, visual_features, terrain_features)
            loss = criterion(outputs.squeeze(), labels)
            
            loss.backward()
            optimizer.step()
            
            train_loss += loss.item() * images.size(0)
            train_samples += images.size(0)
        
        train_loss /= train_samples
        train_losses.append(train_loss)
        
        # Validation phase
        model.eval()
        val_loss = 0.0
        val_samples = 0
        val_predictions = []
        val_targets = []
        
        with torch.no_grad():
            for batch in val_loader:
                images = batch['image'].to(device)
                visual_features = batch['visual_features'].to(device)
                terrain_features = batch['terrain_features'].to(device)
                labels = batch['label'].to(device)
                
                outputs = model(images, visual_features, terrain_features)
                loss = criterion(outputs.squeeze(), labels)
                
                val_loss += loss.item() * images.size(0)
                val_samples += images.size(0)
                
                # Store predictions for metrics
                if outputs.dim() == 0:
                    val_predictions.append(outputs.cpu().item())
                else:
                    val_predictions.extend(outputs.squeeze().cpu().tolist())
                val_targets.extend(labels.cpu().tolist())
        
        val_loss /= val_samples
        val_losses.append(val_loss)
        
        # Calculate additional metrics
        val_mae = mean_absolute_error(val_targets, val_predictions)
        
        # Learning rate scheduling
        scheduler.step(val_loss)
        
        # Save best model
        if val_loss < best_val_loss:
            best_val_loss = val_loss
            torch.save(model.state_dict(), 'best_scenic_model.pth')
        
        print(f'Epoch [{epoch+1}/{num_epochs}], Train Loss: {train_loss:.4f}, Val Loss: {val_loss:.4f}, Val MAE: {val_mae:.4f}')
    
    return train_losses, val_losses

# Train the model if we have data
if len(dataset) > 10:  # Need at least some samples
    # Split dataset
    train_size = int(0.8 * len(dataset))
    val_size = len(dataset) - train_size
    train_dataset, val_dataset = torch.utils.data.random_split(dataset, [train_size, val_size])
    
    # Apply training transforms to train set
    train_dataset.dataset.transform = transform_train
    
    # Create data loaders
    train_loader = DataLoader(train_dataset, batch_size=8, shuffle=True, num_workers=2)
    val_loader = DataLoader(val_dataset, batch_size=8, shuffle=False, num_workers=2)
    
    print(f"Training with {len(train_dataset)} samples, validating with {len(val_dataset)} samples")
    
    # Train the model
    train_losses, val_losses = train_model(model, train_loader, val_loader, num_epochs=15, lr=0.001)
    
    # Plot training history
    plt.figure(figsize=(10, 4))
    
    plt.subplot(1, 2, 1)
    plt.plot(train_losses, label='Training Loss')
    plt.plot(val_losses, label='Validation Loss')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.title('Training History')
    plt.legend()
    plt.grid(True)
    
    # Final evaluation
    model.eval()
    all_predictions = []
    all_targets = []
    
    with torch.no_grad():
        for batch in val_loader:
            images = batch['image'].to(device)
            visual_features = batch['visual_features'].to(device)
            terrain_features = batch['terrain_features'].to(device)
            labels = batch['label'].to(device)
            
            outputs = model(images, visual_features, terrain_features)
            
            if outputs.dim() == 0:
                all_predictions.append(outputs.cpu().item())
            else:
                all_predictions.extend(outputs.squeeze().cpu().tolist())
            all_targets.extend(labels.cpu().tolist())
    
    plt.subplot(1, 2, 2)
    plt.scatter(all_targets, all_predictions, alpha=0.6)
    plt.plot([0, 10], [0, 10], 'r--', alpha=0.8)
    plt.xlabel('True Scenic Score')
    plt.ylabel('Predicted Scenic Score')
    plt.title('Predictions vs Targets')
    plt.grid(True)
    
    plt.tight_layout()
    plt.show()
    
    final_mse = mean_squared_error(all_targets, all_predictions)
    final_mae = mean_absolute_error(all_targets, all_predictions)
    
    print(f"\nFinal Validation Metrics:")
    print(f"MSE: {final_mse:.4f}")
    print(f"MAE: {final_mae:.4f}")
    print(f"RMSE: {np.sqrt(final_mse):.4f}")
    
    print("\nModel training completed! Best model saved as 'best_scenic_model.pth'")
    
else:
    print("Not enough data for training. Need to download more tiles first.")
```

```python
def predict_scenic_score(model, image_path, terrain_path, visual_scaler, terrain_scaler):
    """Predict scenic score for a single tile"""
    model.eval()
    
    # Load and process image
    image = Image.open(image_path).convert('RGB')
    image_tensor = transform_val(image).unsqueeze(0).to(device)
    
    # Extract visual features
    img_array = np.array(image)
    visual_features = enhanced_visual_analysis(img_array)
    visual_vector = np.array([
        visual_features['water_ratio'],
        visual_features['vegetation_ratio'],
        visual_features['snow_ratio'],
        visual_features['rock_ratio'],
        visual_features['urban_ratio'],
        visual_features['edge_density'],
        visual_features['texture_variation'],
        visual_features['color_diversity']
    ]).reshape(1, -1)
    
    # Extract terrain features
    terrain_img = np.array(Image.open(terrain_path))
    elevation = decode_terrain_rgb(terrain_img)
    terrain_features = calculate_terrain_features(elevation)
    terrain_vector = np.array([
        terrain_features['elevation_mean'],
        terrain_features['elevation_std'],
        terrain_features['elevation_range'],
        terrain_features['slope_mean'],
        terrain_features['slope_std'],
        terrain_features['aspect_variation'],
        terrain_features['roughness'],
        terrain_features['tpi_std']
    ]).reshape(1, -1)
    
    # Normalize features
    visual_normalized = visual_scaler.transform(visual_vector)
    terrain_normalized = terrain_scaler.transform(terrain_vector)
    
    # Convert to tensors
    visual_tensor = torch.tensor(visual_normalized, dtype=torch.float32).to(device)
    terrain_tensor = torch.tensor(terrain_normalized, dtype=torch.float32).to(device)
    
    # Predict
    with torch.no_grad():
        prediction = model(image_tensor, visual_tensor, terrain_tensor)
    
    return prediction.cpu().item(), visual_features, terrain_features

# Test prediction on a sample if we have trained data
if len(dataset) > 0 and os.path.exists('best_scenic_model.pth'):
    # Load best model
    model.load_state_dict(torch.load('best_scenic_model.pth', map_location=device))
    
    # Test on a sample from dataset
    sample_idx = 0
    sat_file, terrain_file = dataset.tile_pairs[sample_idx]
    sat_path = os.path.join(TILES_DIR, sat_file)
    terrain_path = os.path.join(TERRAIN_DIR, terrain_file)
    
    predicted_score, visual_features, terrain_features = predict_scenic_score(
        model, sat_path, terrain_path, 
        dataset.visual_scaler, dataset.terrain_scaler
    )
    
    actual_score = dataset.labels[sample_idx]
    
    print(f"\nSample Prediction for {sat_file}:")
    print(f"Predicted Score: {predicted_score:.2f}")
    print(f"Actual Score: {actual_score:.2f}")
    print(f"Difference: {abs(predicted_score - actual_score):.2f}")
    
    print(f"\nVisual Features:")
    for key, value in visual_features.items():
        print(f"  {key}: {value:.4f}")
    
    print(f"\nTerrain Features:")
    for key, value in terrain_features.items():
        print(f"  {key}: {value:.4f}")
        
    # Visualize the sample
    fig, axes = plt.subplots(1, 3, figsize=(15, 5))
    
    # Original image
    img = Image.open(sat_path)
    axes[0].imshow(img)
    axes[0].set_title(f'Satellite Image\nPredicted: {predicted_score:.2f}\nActual: {actual_score:.2f}')
    axes[0].axis('off')
    
    # Terrain visualization
    terrain_img = np.array(Image.open(terrain_path))
    elevation = decode_terrain_rgb(terrain_img)
    
    im1 = axes[1].imshow(elevation, cmap='terrain')
    axes[1].set_title(f'Elevation\n{elevation.min():.0f}-{elevation.max():.0f}m')
    axes[1].axis('off')
    plt.colorbar(im1, ax=axes[1], shrink=0.8)
    
    # Feature importance
    feature_names = ['Water', 'Vegetation', 'Snow', 'Rock', 'Urban', 'Edges', 'Texture', 'Color']
    feature_values = [
        visual_features['water_ratio'],
        visual_features['vegetation_ratio'],
        visual_features['snow_ratio'],
        visual_features['rock_ratio'],
        visual_features['urban_ratio'],
        visual_features['edge_density'],
        visual_features['texture_variation'] / 100,  # Scale for visualization
        visual_features['color_diversity']
    ]
    
    bars = axes[2].bar(range(len(feature_names)), feature_values)
    axes[2].set_xticks(range(len(feature_names)))
    axes[2].set_xticklabels(feature_names, rotation=45, ha='right')
    axes[2].set_title('Visual Features')
    axes[2].set_ylabel('Feature Value')
    
    plt.tight_layout()
    plt.show()
    
else:
    print("No trained model available for prediction.")
```

## Model Summary

This complete scenic drive rating model combines:

### ✅ **Enhanced Data Collection**
- **Higher resolution** tiles (zoom 14) from Denver-Rockies region
- **Advanced visual analysis** with multi-color space water/vegetation detection
- **Comprehensive terrain features** including elevation, slope, curvature, and topographic indices

### 🧠 **Advanced Model Architecture**
- **HuggingFace Vision Transformer** with fallback to EfficientNet
- **Multi-modal fusion** combining visual, visual features, and terrain data
- **Balanced scoring** (60% visual, 40% terrain) to prevent terrain dominance

### 🎯 **Key Improvements**
- **Water bodies** are correctly identified as scenic (not misclassified as plains)
- **Terrain complexity** is properly weighted with visual appeal
- **Robust feature extraction** works with or without OpenCV/HuggingFace
- **Complete training pipeline** with validation and model saving

### 🚀 **Usage**
1. **Download tiles** by running the download cell
2. **Process features** and generate scenic scores
3. **Train model** with the complete dataset
4. **Make predictions** on new tiles for scenic route planning

The model is now ready to provide accurate scenic ratings for driving route optimization!
