# Improved Scenic Drive Rating Model

This enhanced version addresses the key issues:
1. **Better Region**: Denver-Rockies area for diverse terrain (mountains, plains, foothills)
2. **Higher Resolution**: Zoom level 14 for more detailed tiles
3. **Advanced Visual Features**: Better water/vegetation detection using color spaces
4. **HuggingFace Integration**: Pre-trained vision transformer for robust visual features
5. **Balanced Scoring**: Proper weighting between visual and terrain features

**New Region**: Denver Metro to Rocky Mountain National Park (diverse terrain)
**Higher Zoom**: Level 14 (4x more detail than level 12)

```python
# Enhanced Visual Feature Detection
def enhanced_visual_analysis(img_array: np.ndarray) -> Dict[str, float]:
    """Advanced visual feature detection using multiple color spaces"""
    h, w, _ = img_array.shape
    
    # Convert to different color spaces for better feature detection
    hsv = cv2.cvtColor(img_array, cv2.COLOR_RGB2HSV)
    lab = cv2.cvtColor(img_array, cv2.COLOR_RGB2LAB)
    
    features = {}
    
    # Water detection (multiple methods)
    # Method 1: Blue dominance in RGB
    blue_dominant = (
        (img_array[:,:,2] > img_array[:,:,1] + 10) & 
        (img_array[:,:,2] > img_array[:,:,0] + 10) &
        (img_array[:,:,2] > 80)
    )
    
    # Method 2: Low saturation, mid-high value (lakes/reservoirs)
    water_hsv = (
        (hsv[:,:,1] < 100) &  # Low saturation
        (hsv[:,:,2] > 50) &   # Mid-high value
        (hsv[:,:,2] < 200)    # Not too bright (avoid clouds)
    )
    
    # Method 3: Dark blue hues (rivers, deep water)
    dark_blue = (
        ((hsv[:,:,0] > 90) & (hsv[:,:,0] < 130)) &  # Blue hue range
        (hsv[:,:,1] > 50) &   # Some saturation
        (hsv[:,:,2] > 30)     # Not completely dark
    )
    
    water_combined = blue_dominant | water_hsv | dark_blue
    features['water_ratio'] = np.sum(water_combined) / (h * w)
    
    # Vegetation detection (enhanced)
    # Method 1: Green dominance
    green_dominant = (
        (img_array[:,:,1] > img_array[:,:,0] + 15) & 
        (img_array[:,:,1] > img_array[:,:,2] + 15) &
        (img_array[:,:,1] > 60)
    )
    
    # Method 2: NDVI-like calculation (NIR approximation)
    # Use green-red ratio as rough NDVI proxy
    green = img_array[:,:,1].astype(np.float32)
    red = img_array[:,:,0].astype(np.float32)
    ndvi_proxy = (green - red) / (green + red + 1e-8)
    vegetation_ndvi = ndvi_proxy > 0.1
    
    vegetation_combined = green_dominant | vegetation_ndvi
    features['vegetation_ratio'] = np.sum(vegetation_combined) / (h * w)
    
    # Snow/ice detection
    snow_mask = (
        (img_array[:,:,0] > 200) & 
        (img_array[:,:,1] > 200) & 
        (img_array[:,:,2] > 200) &
        (np.abs(img_array[:,:,0] - img_array[:,:,1]) < 20) &
        (np.abs(img_array[:,:,1] - img_array[:,:,2]) < 20)
    )
    features['snow_ratio'] = np.sum(snow_mask) / (h * w)
    
    # Rock/bare earth (brown/tan colors)
    rock_mask = (
        (hsv[:,:,0] > 10) & (hsv[:,:,0] < 30) &  # Brown/tan hues
        (hsv[:,:,1] > 30) & (hsv[:,:,1] < 150) & # Moderate saturation
        (hsv[:,:,2] > 50) & (hsv[:,:,2] < 180)   # Moderate brightness
    )
    features['rock_ratio'] = np.sum(rock_mask) / (h * w)
    
    # Urban/developed areas (gray, regular patterns)
    urban_mask = (
        (np.abs(img_array[:,:,0] - img_array[:,:,1]) < 25) &
        (np.abs(img_array[:,:,1] - img_array[:,:,2]) < 25) &
        (np.mean(img_array, axis=2) > 60) &
        (np.mean(img_array, axis=2) < 180)
    )
    features['urban_ratio'] = np.sum(urban_mask) / (h * w)
    
    # Texture analysis
    gray = cv2.cvtColor(img_array, cv2.COLOR_RGB2GRAY)
    
    # Edge density (terrain variation)
    edges = cv2.Canny(gray, 50, 150)
    features['edge_density'] = np.sum(edges > 0) / (h * w)
    
    # Texture variation (local standard deviation)
    kernel = cv2.getGaussianKernel(9, 2)
    kernel = kernel @ kernel.T
    mean_filtered = cv2.filter2D(gray.astype(np.float32), -1, kernel)
    sq_filtered = cv2.filter2D((gray.astype(np.float32))**2, -1, kernel)
    local_variance = sq_filtered - mean_filtered**2
    features['texture_variation'] = np.mean(np.sqrt(np.maximum(local_variance, 0)))
    
    # Color diversity (number of distinct colors)
    unique_colors = len(np.unique(img_array.reshape(-1, 3), axis=0))
    features['color_diversity'] = unique_colors / (h * w)  # Normalize
    
    return features

# Test enhanced visual analysis
test_files = [f for f in os.listdir(TILES_DIR) if f.endswith('.png')]
if test_files:
    test_file = test_files[0]
    test_img = np.array(Image.open(os.path.join(TILES_DIR, test_file)))
    features = enhanced_visual_analysis(test_img)
    
    print(f"Enhanced visual features for {test_file}:")
    for feature, value in features.items():
        print(f"  {feature}: {value:.4f}")
    
    # Visualize feature detection
    fig, axes = plt.subplots(2, 3, figsize=(15, 10))
    axes = axes.flatten()
    
    axes[0].imshow(test_img)
    axes[0].set_title('Original Image')
    axes[0].axis('off')
    
    # Show detected features as masks
    h, w, _ = test_img.shape
    hsv = cv2.cvtColor(test_img, cv2.COLOR_RGB2HSV)
    
    # Water mask
    water_mask = (
        (test_img[:,:,2] > test_img[:,:,1] + 10) & 
        (test_img[:,:,2] > test_img[:,:,0] + 10) &
        (test_img[:,:,2] > 80)
    )
    axes[1].imshow(water_mask, cmap='Blues')
    axes[1].set_title(f'Water Detection ({features["water_ratio"]:.3f})')
    axes[1].axis('off')
    
    # Vegetation mask
    veg_mask = (
        (test_img[:,:,1] > test_img[:,:,0] + 15) & 
        (test_img[:,:,1] > test_img[:,:,2] + 15) &
        (test_img[:,:,1] > 60)
    )
    axes[2].imshow(veg_mask, cmap='Greens')
    axes[2].set_title(f'Vegetation Detection ({features["vegetation_ratio"]:.3f})')
    axes[2].axis('off')
    
    # Edge detection
    gray = cv2.cvtColor(test_img, cv2.COLOR_RGB2GRAY)
    edges = cv2.Canny(gray, 50, 150)
    axes[3].imshow(edges, cmap='gray')
    axes[3].set_title(f'Edges ({features["edge_density"]:.3f})')
    axes[3].axis('off')
    
    # Show HSV for analysis
    axes[4].imshow(hsv)
    axes[4].set_title('HSV Color Space')
    axes[4].axis('off')
    
    # Combined analysis
    axes[5].text(0.1, 0.8, f"Water: {features['water_ratio']:.3f}", transform=axes[5].transAxes, fontsize=12)
    axes[5].text(0.1, 0.7, f"Vegetation: {features['vegetation_ratio']:.3f}", transform=axes[5].transAxes, fontsize=12)
    axes[5].text(0.1, 0.6, f"Snow: {features['snow_ratio']:.3f}", transform=axes[5].transAxes, fontsize=12)
    axes[5].text(0.1, 0.5, f"Rock: {features['rock_ratio']:.3f}", transform=axes[5].transAxes, fontsize=12)
    axes[5].text(0.1, 0.4, f"Urban: {features['urban_ratio']:.3f}", transform=axes[5].transAxes, fontsize=12)
    axes[5].text(0.1, 0.3, f"Texture: {features['texture_variation']:.1f}", transform=axes[5].transAxes, fontsize=12)
    axes[5].set_title('Feature Summary')
    axes[5].axis('off')
    
    plt.tight_layout()
    plt.show()
else:
    print("No tiles found. Run download first.")
```

```python
# Enhanced terrain processing (same as before but optimized)
def decode_terrain_rgb(terrain_img: np.ndarray) -> np.ndarray:
    """Decode terrain-rgb image to elevation in meters"""
    r, g, b = terrain_img[:, :, 0], terrain_img[:, :, 1], terrain_img[:, :, 2]
    elevation = -10000 + ((r.astype(np.float32) * 256 * 256 + 
                          g.astype(np.float32) * 256 + 
                          b.astype(np.float32)) * 0.1)
    return elevation

def calculate_enhanced_terrain_features(elevation: np.ndarray) -> Dict[str, float]:
    """Calculate comprehensive terrain features"""
    # Basic elevation statistics
    elevation_mean = np.mean(elevation)
    elevation_std = np.std(elevation)
    elevation_range = np.max(elevation) - np.min(elevation)
    elevation_percentile_range = np.percentile(elevation, 95) - np.percentile(elevation, 5)
    
    # Slope analysis
    dy, dx = np.gradient(elevation)
    slope = np.sqrt(dx**2 + dy**2)
    slope_mean = np.mean(slope)
    slope_std = np.std(slope)
    slope_max = np.max(slope)
    
    # Aspect (slope direction) analysis
    aspect = np.arctan2(dy, dx)
    aspect_variation = np.std(aspect)
    
    # Terrain roughness (multiple measures)
    laplacian = ndimage.laplace(elevation)
    roughness = np.std(laplacian)
    
    # Terrain Ruggedness Index (TRI)
    tri = np.sqrt(np.sum(np.gradient(elevation)**2, axis=0))
    tri_mean = np.mean(tri)
    
    # Profile curvature (curvature in direction of steepest descent)
    dxx = ndimage.sobel(dx, axis=1)
    dyy = ndimage.sobel(dy, axis=0)
    dxy = ndimage.sobel(dx, axis=0)
    
    slope_safe = np.where(slope > 1e-6, slope, 1e-6)
    profile_curvature = (dxx * dx**2 + 2 * dxy * dx * dy + dyy * dy**2) / (slope_safe**3)
    profile_curvature_mean = np.mean(np.abs(profile_curvature))
    
    # Plan curvature (curvature perpendicular to slope direction)
    plan_curvature = (dxx * dy**2 - 2 * dxy * dx * dy + dyy * dx**2) / (slope_safe**2)
    plan_curvature_mean = np.mean(np.abs(plan_curvature))
    
    # Terrain Position Index (TPI) - multiple scales
    kernel_small = np.ones((3, 3)) / 9
    kernel_large = np.ones((7, 7)) / 49
    
    tpi_small = elevation - ndimage.convolve(elevation, kernel_small, mode='constant')
    tpi_large = elevation - ndimage.convolve(elevation, kernel_large, mode='constant')
    
    tpi_small_std = np.std(tpi_small)
    tpi_large_std = np.std(tpi_large)
    
    # Valley/ridge detection
    valleys = np.sum(tpi_small < -np.std(tpi_small)) / tpi_small.size
    ridges = np.sum(tpi_small > np.std(tpi_small)) / tpi_small.size
    
    return {
        'elevation_mean': elevation_mean,
        'elevation_std': elevation_std,
        'elevation_range': elevation_range,
        'elevation_percentile_range': elevation_percentile_range,
        'slope_mean': slope_mean,
        'slope_std': slope_std,
        'slope_max': slope_max,
        'aspect_variation': aspect_variation,
        'roughness': roughness,
        'tri_mean': tri_mean,
        'profile_curvature_mean': profile_curvature_mean,
        'plan_curvature_mean': plan_curvature_mean,
        'tpi_small_std': tpi_small_std,
        'tpi_large_std': tpi_large_std,
        'valley_ratio': valleys,
        'ridge_ratio': ridges
    }

# Test enhanced terrain features
terrain_files = [f for f in os.listdir(TERRAIN_DIR) if f.endswith('_terrain.png')]
if terrain_files:
    test_terrain = terrain_files[0]
    terrain_img = np.array(Image.open(os.path.join(TERRAIN_DIR, test_terrain)))
    elevation = decode_terrain_rgb(terrain_img)
    terrain_features = calculate_enhanced_terrain_features(elevation)
    
    print(f"Enhanced terrain features for {test_terrain}:")
    for feature, value in terrain_features.items():
        print(f"  {feature}: {value:.4f}")
    
    # Visualize terrain analysis
    fig, axes = plt.subplots(2, 3, figsize=(15, 10))
    axes = axes.flatten()
    
    # Original terrain RGB
    axes[0].imshow(terrain_img)
    axes[0].set_title('Terrain RGB')
    axes[0].axis('off')
    
    # Elevation
    im1 = axes[1].imshow(elevation, cmap='terrain')
    axes[1].set_title(f'Elevation\n{elevation.min():.0f}-{elevation.max():.0f}m')
    axes[1].axis('off')
    plt.colorbar(im1, ax=axes[1], shrink=0.6)
    
    # Slope
    dy, dx = np.gradient(elevation)
    slope = np.sqrt(dx**2 + dy**2)
    im2 = axes[2].imshow(slope, cmap='hot')
    axes[2].set_title(f'Slope\nMean: {terrain_features["slope_mean"]:.2f}')
    axes[2].axis('off')
    plt.colorbar(im2, ax=axes[2], shrink=0.6)
    
    # TPI (terrain position)
    kernel = np.ones((5, 5)) / 25
    tpi = elevation - ndimage.convolve(elevation, kernel, mode='constant')
    im3 = axes[3].imshow(tpi, cmap='RdBu_r', vmin=-20, vmax=20)
    axes[3].set_title('Terrain Position Index\n(Red=Ridge, Blue=Valley)')
    axes[3].axis('off')
    plt.colorbar(im3, ax=axes[3], shrink=0.6)
    
    # Profile curvature
    slope_safe = np.where(slope > 1e-6, slope, 1e-6)
    dxx = ndimage.sobel(dx, axis=1)
    dyy = ndimage.sobel(dy, axis=0)
    dxy = ndimage.sobel(dx, axis=0)
    profile_curvature = (dxx * dx**2 + 2 * dxy * dx * dy + dyy * dy**2) / (slope_safe**3)
    im4 = axes[4].imshow(profile_curvature, cmap='RdBu_r', vmin=-0.01, vmax=0.01)
    axes[4].set_title('Profile Curvature\n(Convex/Concave)')
    axes[4].axis('off')
    plt.colorbar(im4, ax=axes[4], shrink=0.6)
    
    # Feature summary
    axes[5].text(0.05, 0.9, f"Elev Range: {terrain_features['elevation_range']:.0f}m", transform=axes[5].transAxes)
    axes[5].text(0.05, 0.8, f"Slope Mean: {terrain_features['slope_mean']:.2f}", transform=axes[5].transAxes)
    axes[5].text(0.05, 0.7, f"Roughness: {terrain_features['roughness']:.2f}", transform=axes[5].transAxes)
    axes[5].text(0.05, 0.6, f"TRI: {terrain_features['tri_mean']:.2f}", transform=axes[5].transAxes)
    axes[5].text(0.05, 0.5, f"Valleys: {terrain_features['valley_ratio']:.3f}", transform=axes[5].transAxes)
    axes[5].text(0.05, 0.4, f"Ridges: {terrain_features['ridge_ratio']:.3f}", transform=axes[5].transAxes)
    axes[5].set_title('Terrain Summary')
    axes[5].axis('off')
    
    plt.tight_layout()
    plt.show()
else:
    print("No terrain files found. Run download first.")
```

```python
# HuggingFace Vision Model Integration
class HuggingFaceVisualEncoder(nn.Module):
    def __init__(self, model_name="microsoft/swin-tiny-patch4-window7-224"):
        super().__init__()
        
        # Load pre-trained vision transformer
        self.config = AutoConfig.from_pretrained(model_name)
        self.model = AutoModel.from_pretrained(model_name)
        self.processor = AutoImageProcessor.from_pretrained(model_name)
        
        # Get feature dimension
        self.feature_dim = self.config.hidden_size
        
        # Freeze backbone (optional - can fine-tune if needed)
        for param in self.model.parameters():
            param.requires_grad = False
            
        # Add trainable projection layer
        self.projection = nn.Sequential(
            nn.Linear(self.feature_dim, 512),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(512, 256)
        )
        
    def forward(self, pixel_values):
        # Get features from vision transformer
        outputs = self.model(pixel_values=pixel_values)
        
        # Use pooled output (global features)
        if hasattr(outputs, 'pooler_output') and outputs.pooler_output is not None:
            features = outputs.pooler_output
        else:
            # Fallback: global average pool of last hidden state
            features = outputs.last_hidden_state.mean(dim=1)
        
        # Project to desired dimension
        projected_features = self.projection(features)
        
        return projected_features

# Advanced Multi-Modal Model
class AdvancedScenicCNN(nn.Module):
    def __init__(self, visual_features_dim=9, terrain_features_dim=16, use_huggingface=True):
        super().__init__()
        
        self.use_huggingface = use_huggingface
        
        if use_huggingface:
            # HuggingFace vision encoder
            self.visual_encoder = HuggingFaceVisualEncoder()
            visual_feature_dim = 256  # From projection layer
        else:
            # Traditional CNN backbone
            self.visual_backbone = models.efficientnet_b0(pretrained=True)
            visual_feature_dim = self.visual_backbone.classifier[1].in_features
            self.visual_backbone.classifier = nn.Identity()
        
        # Enhanced visual features processor
        self.visual_features_processor = nn.Sequential(
            nn.Linear(visual_features_dim, 32),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(32, 16)
        )
        
        # Terrain features processor
        self.terrain_processor = nn.Sequential(
            nn.Linear(terrain_features_dim, 64),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(64, 32),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(32, 16)
        )
        
        # Multi-head attention for feature fusion
        self.attention = nn.MultiheadAttention(
            embed_dim=16,
            num_heads=4,
            batch_first=True
        )
        
        # Final fusion and prediction layers
        fusion_input_dim = visual_feature_dim + 16 + 16  # visual + visual_features + terrain
        
        self.fusion_layers = nn.Sequential(
            nn.Linear(fusion_input_dim, 512),
            nn.ReLU(),
            nn.Dropout(0.4),
            nn.Linear(512, 256),
            nn.ReLU(),
            nn.Dropout(0.4),
            nn.Linear(256, 128),
            nn.ReLU(),
            nn.Linear(128, 64),
            nn.ReLU(),
            nn.Linear(64, 1),
            nn.Sigmoid()
        )
        
        # Feature importance weights (learned)
        self.feature_weights = nn.Parameter(torch.tensor([1.0, 1.0, 1.0]))  # visual, visual_features, terrain
        
    def forward(self, images, visual_features, terrain_features):
        # Process image through vision model
        if self.use_huggingface:
            # HuggingFace expects specific input format
            visual_encoded = self.visual_encoder(images)
        else:
            visual_encoded = self.visual_backbone(images)
        
        # Process enhanced visual features
        visual_features_processed = self.visual_features_processor(visual_features)
        
        # Process terrain features
        terrain_processed = self.terrain_processor(terrain_features)
        
        # Apply attention between visual features and terrain
        feature_stack = torch.stack([
            visual_features_processed,
            terrain_processed
        ], dim=1)  # Shape: [batch, 2, 16]
        
        attended_features, _ = self.attention(
            feature_stack, feature_stack, feature_stack
        )
        
        # Flatten attended features
        attended_flat = attended_features.view(attended_features.size(0), -1)
        
        # Apply learned feature weights
        weights = torch.softmax(self.feature_weights, dim=0)
        
        weighted_visual = visual_encoded * weights[0]
        weighted_attended = attended_flat * weights[1]
        
        # Combine all features
        combined_features = torch.cat([
            weighted_visual,
            weighted_attended
        ], dim=1)
        
        # Final prediction
        output = self.fusion_layers(combined_features) * 10  # Scale to 0-10
        
        return output

# Test model initialization
try:
    model = AdvancedScenicCNN(visual_features_dim=9, terrain_features_dim=16, use_huggingface=True)
    model = model.to(device)
    print(f"Advanced model initialized with HuggingFace vision encoder")
    print(f"Model parameters: {sum(p.numel() for p in model.parameters()):,}")
    print(f"Trainable parameters: {sum(p.numel() for p in model.parameters() if p.requires_grad):,}")
except Exception as e:
    print(f"Error with HuggingFace model: {e}")
    print("Falling back to traditional CNN...")
    model = AdvancedScenicCNN(visual_features_dim=9, terrain_features_dim=16, use_huggingface=False)
    model = model.to(device)
    print(f"Traditional CNN model initialized")
    print(f"Model parameters: {sum(p.numel() for p in model.parameters()):,}")
```

```python
# Enhanced Dataset with Balanced Scoring
class AdvancedScenicDataset(Dataset):
    def __init__(self, tiles_dir: str, terrain_dir: str, transform=None, create_labels=True):
        self.tiles_dir = tiles_dir
        self.terrain_dir = terrain_dir
        self.transform = transform
        
        # Find matching pairs
        self.tile_pairs = self._find_matching_tiles()
        print(f"Found {len(self.tile_pairs)} matching satellite-terrain tile pairs")
        
        if create_labels and len(self.tile_pairs) > 0:
            self.labels, self.visual_features, self.terrain_features = self._generate_balanced_scores()
            
            # Normalize features
            self.visual_scaler = StandardScaler()
            self.terrain_scaler = StandardScaler()
            
            self.visual_features_normalized = self.visual_scaler.fit_transform(self.visual_features)
            self.terrain_features_normalized = self.terrain_scaler.fit_transform(self.terrain_features)
        else:
            # Initialize empty arrays
            n_samples = len(self.tile_pairs)
            self.labels = [5.0] * n_samples  # Default neutral score
            self.visual_features = np.zeros((n_samples, 9))
            self.terrain_features = np.zeros((n_samples, 16))
            self.visual_features_normalized = self.visual_features
            self.terrain_features_normalized = self.terrain_features
    
    def _find_matching_tiles(self) -> List[Tuple[str, str]]:
        """Find satellite tiles with corresponding terrain tiles"""
        if not os.path.exists(self.tiles_dir) or not os.path.exists(self.terrain_dir):
            return []
            
        satellite_files = [f for f in os.listdir(self.tiles_dir) if f.endswith('.png')]
        terrain_files = [f for f in os.listdir(self.terrain_dir) if f.endswith('_terrain.png')]
        
        pairs = []
        for sat_file in satellite_files:
            base_name = sat_file.replace('.png', '')
            terrain_file = f"{base_name}_terrain.png"
            
            if terrain_file in terrain_files:
                pairs.append((sat_file, terrain_file))
        
        return pairs
    
    def _generate_balanced_scores(self) -> Tuple[List[float], np.ndarray, np.ndarray]:
        """Generate balanced scenic scores with proper visual-terrain weighting"""
        scores = []
        visual_features_list = []
        terrain_features_list = []
        
        for sat_file, terrain_file in self.tile_pairs:
            try:
                # Load and analyze satellite image
                sat_path = os.path.join(self.tiles_dir, sat_file)
                sat_img = np.array(Image.open(sat_path).convert('RGB'))
                visual_features = enhanced_visual_analysis(sat_img)
                
                # Load and analyze terrain
                terrain_path = os.path.join(self.terrain_dir, terrain_file)
                terrain_img = np.array(Image.open(terrain_path))
                elevation = decode_terrain_rgb(terrain_img)
                terrain_features = calculate_enhanced_terrain_features(elevation)
                
                # Calculate balanced scenic score
                visual_score = self._calculate_visual_score(visual_features)
                terrain_score = self._calculate_terrain_score(terrain_features)
                
                # Weighted combination (60% visual, 40% terrain)
                combined_score = 0.6 * visual_score + 0.4 * terrain_score
                scores.append(min(10.0, max(0.0, combined_score)))
                
                # Store features
                visual_vector = [
                    visual_features['water_ratio'],
                    visual_features['vegetation_ratio'],
                    visual_features['snow_ratio'],
                    visual_features['rock_ratio'],
                    visual_features['urban_ratio'],
                    visual_features['edge_density'],
                    visual_features['texture_variation'],
                    visual_features['color_diversity'],
                    visual_score  # Include calculated visual score
                ]
                visual_features_list.append(visual_vector)
                
                terrain_vector = [
                    terrain_features['elevation_mean'],
                    terrain_features['elevation_std'],
                    terrain_features['elevation_range'],
                    terrain_features['elevation_percentile_range'],
                    terrain_features['slope_mean'],
                    terrain_features['slope_std'],
                    terrain_features['slope_max'],
                    terrain_features['aspect_variation'],
                    terrain_features['roughness'],
                    terrain_features['tri_mean'],
                    terrain_features['profile_curvature_mean'],
                    terrain_features['plan_curvature_mean'],
                    terrain_features['tpi_small_std'],
                    terrain_features['tpi_large_std'],
                    terrain_features['valley_ratio'],
                    terrain_features['ridge_ratio']
                ]
                terrain_features_list.append(terrain_vector)
                
            except Exception as e:
                print(f"Error processing {sat_file}: {e}")
                # Add default values
                scores.append(5.0)
                visual_features_list.append([0.0] * 9)
                terrain_features_list.append([0.0] * 16)
        
        return scores, np.array(visual_features_list), np.array(terrain_features_list)
    
    def _calculate_visual_score(self, features: Dict[str, float]) -> float:
        """Calculate visual scenic score with improved logic"""
        score = 2.0  # Base score
        
        # Water bodies are highly scenic
        water_score = min(3.0, features['water_ratio'] * 20)  # Max 3 points
        score += water_score
        
        # Vegetation adds beauty
        veg_score = min(2.0, features['vegetation_ratio'] * 8)  # Max 2 points
        score += veg_score
        
        # Snow/ice can be scenic
        snow_score = min(1.5, features['snow_ratio'] * 10)  # Max 1.5 points
        score += snow_score
        
        # Rock formations add character
        rock_score = min(1.0, features['rock_ratio'] * 5)  # Max 1 point
        score += rock_score
        
        # Edge density indicates interesting features
        edge_score = min(1.5, features['edge_density'] * 15)  # Max 1.5 points
        score += edge_score
        
        # Texture variation adds visual interest
        texture_score = min(1.0, features['texture_variation'] / 50)  # Max 1 point
        score += texture_score
        
        # Color diversity
        color_score = min(0.5, features['color_diversity'] * 10)  # Max 0.5 points
        score += color_score
        
        # Urban penalty
        urban_penalty = features['urban_ratio'] * 3  # Subtract up to 3 points
        score -= urban_penalty
        
        return max(0.0, score)
    
    def _calculate_terrain_score(self, features: Dict[str, float]) -> float:
        """Calculate terrain scenic score"""
        score = 1.0  # Base score
        
        # Elevation variation is scenic
        elev_range = features['elevation_range']
        if elev_range > 500:  # Mountains
            score += 3.0
        elif elev_range > 200:  # Hills
            score += 2.0
        elif elev_range > 100:  # Rolling terrain
            score += 1.0
        # else: flat (no bonus)
        
        # Optimal slope range
        slope_mean = features['slope_mean']
        if 2 < slope_mean < 15:  # Moderate slopes
            score += 2.0
        elif 1 < slope_mean < 25:  # Gentle to steep
            score += 1.0
        # Very flat or very steep get no bonus
        
        # Terrain complexity
        roughness_score = min(1.5, features['roughness'] / 20)
        score += roughness_score
        
        # Ridges and valleys add interest
        topo_features = features['ridge_ratio'] + features['valley_ratio']
        topo_score = min(1.0, topo_features * 10)
        score += topo_score
        
        # Curvature adds visual interest
        curvature_score = min(1.0, (features['profile_curvature_mean'] + features['plan_curvature_mean']) * 500)
        score += curvature_score
        
        return max(0.0, score)
    
    def __len__(self):
        return len(self.tile_pairs)
    
    def __getitem__(self, idx):
        if idx >= len(self.tile_pairs):
            raise IndexError(f"Index {idx} out of range for dataset of size {len(self.tile_pairs)}")
            
        sat_file, terrain_file = self.tile_pairs[idx]
        
        # Load satellite image
        sat_path = os.path.join(self.tiles_dir, sat_file)
        image = Image.open(sat_path).convert('RGB')
        
        if self.transform:
            image = self.transform(image)
        
        return {
            'image': image,
            'visual_features': torch.tensor(self.visual_features_normalized[idx], dtype=torch.float32),
            'terrain_features': torch.tensor(self.terrain_features_normalized[idx], dtype=torch.float32),
            'label': torch.tensor(self.labels[idx], dtype=torch.float32),
            'sat_filename': sat_file,
            'terrain_filename': terrain_file
        }

# Create enhanced dataset
transform_train = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(0.3),
    transforms.RandomRotation(15),
    transforms.ColorJitter(brightness=0.3, contrast=0.3, saturation=0.2, hue=0.1),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])

transform_val = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])

# Create dataset
dataset = AdvancedScenicDataset(TILES_DIR, TERRAIN_DIR, transform=transform_val)

if len(dataset) > 0:
    print(f"\nLoaded {len(dataset)} tile pairs")
    print(f"Scenic score range: {min(dataset.labels):.2f} - {max(dataset.labels):.2f}")
    print(f"Mean scenic score: {np.mean(dataset.labels):.2f}")
    print(f"Visual features shape: {dataset.visual_features.shape}")
    print(f"Terrain features shape: {dataset.terrain_features.shape}")
    
    # Show score distribution
    plt.figure(figsize=(12, 4))
    
    plt.subplot(1, 2, 1)
    plt.hist(dataset.labels, bins=20, edgecolor='black', alpha=0.7)
    plt.xlabel('Scenic Score')
    plt.ylabel('Frequency')
    plt.title('Enhanced Scenic Score Distribution')
    plt.grid(True, alpha=0.3)
    
    plt.subplot(1, 2, 2)
    # Show correlation between visual and terrain scores
    visual_scores = dataset.visual_features[:, -1]  # Last column is visual score
    plt.scatter(visual_scores, dataset.labels, alpha=0.6)
    plt.xlabel('Visual Score Component')
    plt.ylabel('Total Scenic Score')
    plt.title('Visual vs Total Score')
    plt.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.show()
else:
    print("No matching tile pairs found. Make sure tiles are downloaded.")
```
